```python
# ============================================================
# 04_Exploratory_Data_Analysis.ipynb
# Online Retail Customer Analytics
# ============================================================

# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

# Data manipulation
import pandas as pd
import numpy as np

# Data visualisation
import matplotlib.pyplot as plt
import seaborn as sns

# Statistical analysis
from scipy import stats

# Display settings
pd.set_option("display.max_columns", None)
pd.set_option("display.float_format", lambda x: f"{x:,.2f}")


# ============================================================
# 2. LOAD FEATURE-ENGINEERED DATASETS
# ============================================================

# Transaction-level dataset
transaction_df = pd.read_csv(
    "../Data/retail_transaction_features.csv"
)

# Product-level dataset
product_df = pd.read_csv(
    "../Data/product_features.csv"
)

# Customer-level dataset
customer_df = pd.read_csv(
    "../Data/customer_features.csv"
)

print("Feature-engineered datasets loaded successfully.")


# ============================================================
# 3. EDA DATASET VALIDATION
# ============================================================


# ------------------------------------------------------------
# 3.1 Dataset Dimensions
# ------------------------------------------------------------

dataset_shapes = pd.DataFrame({
    "Dataset": [
        "Transaction",
        "Product",
        "Customer"
    ],
    "Rows": [
        transaction_df.shape[0],
        product_df.shape[0],
        customer_df.shape[0]
    ],
    "Columns": [
        transaction_df.shape[1],
        product_df.shape[1],
        customer_df.shape[1]
    ]
})

display(dataset_shapes)


# ------------------------------------------------------------
# 3.2 Preview Datasets
# ------------------------------------------------------------

print("\nTRANSACTION DATASET")
display(transaction_df.head())

print("\nPRODUCT DATASET")
display(product_df.head())

print("\nCUSTOMER DATASET")
display(customer_df.head())


# ------------------------------------------------------------
# 3.3 Check for Unwanted Index Columns
# ------------------------------------------------------------

datasets = {
    "Transaction": transaction_df,
    "Product": product_df,
    "Customer": customer_df
}

for name, df in datasets.items():

    unnamed_cols = [
        col for col in df.columns
        if str(col).startswith("Unnamed")
    ]

    print(f"{name} unwanted index columns: {unnamed_cols}")


# ------------------------------------------------------------
# 3.4 Dataset Structure
# ------------------------------------------------------------

print("\n--- TRANSACTION DATASET ---")
transaction_df.info()

print("\n--- PRODUCT DATASET ---")
product_df.info()

print("\n--- CUSTOMER DATASET ---")
customer_df.info()


# ------------------------------------------------------------
# 3.5 Column Names
# ------------------------------------------------------------

print("\nTransaction Columns:")
print(transaction_df.columns.tolist())

print("\nProduct Columns:")
print(product_df.columns.tolist())

print("\nCustomer Columns:")
print(customer_df.columns.tolist())


# ------------------------------------------------------------
# 3.6 Data-Type Summary
# ------------------------------------------------------------

def datatype_summary(df):

    return pd.DataFrame({
        "Data_Type": df.dtypes.astype(str),
        "Non_Null_Count": df.notna().sum(),
        "Missing_Count": df.isna().sum(),
        "Unique_Values": df.nunique(dropna=True)
    })


for name, df in datasets.items():

    print(f"\n{name.upper()} DATASET DATA TYPES")

    display(
        datatype_summary(df)
    )


# ------------------------------------------------------------
# 3.7 Missing-Value Validation
# ------------------------------------------------------------

def missing_summary(df):

    summary = pd.DataFrame({
        "Missing_Count": df.isna().sum(),
        "Missing_Percentage":
            (df.isna().mean() * 100).round(2)
    })

    return summary[
        summary["Missing_Count"] > 0
    ].sort_values(
        "Missing_Percentage",
        ascending=False
    )


for name, df in datasets.items():

    print(f"\n{name.upper()} DATASET MISSING VALUES")

    result = missing_summary(df)

    if result.empty:
        print("No missing values detected.")
    else:
        display(result)

    print(
        "Total Missing Values:",
        df.isna().sum().sum()
    )


# ------------------------------------------------------------
# 3.8 Duplicate Validation
# ------------------------------------------------------------

print(
    "Exact Transaction Duplicates:",
    transaction_df.duplicated().sum()
)

print(
    "Exact Product Duplicates:",
    product_df.duplicated().sum()
)

print(
    "Exact Customer Duplicates:",
    customer_df.duplicated().sum()
)


# Validate expected analytical granularity

if "StockCode" in product_df.columns:

    print(
        "Duplicate Product IDs:",
        product_df["StockCode"]
        .duplicated()
        .sum()
    )


if "CustomerID" in customer_df.columns:

    print(
        "Duplicate Customer IDs:",
        customer_df["CustomerID"]
        .duplicated()
        .sum()
    )


# ------------------------------------------------------------
# 3.9 Numerical Feature Identification
# ------------------------------------------------------------

for name, df in datasets.items():

    numerical_cols = (
        df.select_dtypes(include=np.number)
        .columns
        .tolist()
    )

    print(f"\n{name} Numerical Features:")
    print(numerical_cols)


# ------------------------------------------------------------
# 3.10 Infinity Validation
# ------------------------------------------------------------

def infinity_summary(df):

    numeric_df = df.select_dtypes(
        include=np.number
    )

    return pd.DataFrame({
        "Positive_Infinity":
            np.isposinf(numeric_df).sum(),

        "Negative_Infinity":
            np.isneginf(numeric_df).sum(),

        "Total_Infinite":
            np.isinf(numeric_df).sum()
    })


for name, df in datasets.items():

    print(f"\n{name.upper()} INFINITY CHECK")

    display(
        infinity_summary(df)
    )


# ------------------------------------------------------------
# 3.11 Date Validation
# ------------------------------------------------------------

if "InvoiceDate" in transaction_df.columns:

    transaction_df["InvoiceDate"] = pd.to_datetime(
        transaction_df["InvoiceDate"],
        errors="coerce"
    )

    print(
        "Earliest Date:",
        transaction_df["InvoiceDate"].min()
    )

    print(
        "Latest Date:",
        transaction_df["InvoiceDate"].max()
    )

    print(
        "Invalid Dates:",
        transaction_df["InvoiceDate"]
        .isna()
        .sum()
    )


# ------------------------------------------------------------
# 3.12 Temporal Feature Validation
# ------------------------------------------------------------

if "InvoiceYear" in transaction_df.columns:

    print(
        "\nInvoice Years:",
        sorted(
            transaction_df["InvoiceYear"]
            .dropna()
            .unique()
        )
    )


if "InvoiceMonth" in transaction_df.columns:

    print(
        "\nInvoice Months:",
        sorted(
            transaction_df["InvoiceMonth"]
            .dropna()
            .unique()
        )
    )

    print(
        "Invalid Invoice Months:",
        (
            ~transaction_df["InvoiceMonth"]
            .between(1, 12)
        ).sum()
    )


if "MonthName" in transaction_df.columns:

    print(
        "\nMonth Names:",
        transaction_df["MonthName"]
        .dropna()
        .unique()
    )


if "InvoiceDay" in transaction_df.columns:

    print(
        "\nInvoice Day Range:",
        transaction_df["InvoiceDay"].min(),
        "to",
        transaction_df["InvoiceDay"].max()
    )

    print(
        "Invalid Invoice Days:",
        (
            ~transaction_df["InvoiceDay"]
            .between(1, 31)
        ).sum()
    )


if "DayOfWeek" in transaction_df.columns:

    print(
        "\nDays of Week:",
        transaction_df["DayOfWeek"]
        .dropna()
        .unique()
    )


if "InvoiceHour" in transaction_df.columns:

    print(
        "\nInvoice Hours:",
        sorted(
            transaction_df["InvoiceHour"]
            .dropna()
            .unique()
        )
    )

    print(
        "Invalid Hours:",
        (
            ~transaction_df["InvoiceHour"]
            .between(0, 23)
        ).sum()
    )


# ------------------------------------------------------------
# 3.13 Temporal Feature Consistency
# ------------------------------------------------------------

if "InvoiceDate" in transaction_df.columns:

    if "InvoiceYear" in transaction_df.columns:

        print(
            "Year Mismatches:",
            (
                transaction_df["InvoiceYear"]
                != transaction_df["InvoiceDate"].dt.year
            ).sum()
        )

    if "InvoiceMonth" in transaction_df.columns:

        print(
            "Month Mismatches:",
            (
                transaction_df["InvoiceMonth"]
                != transaction_df["InvoiceDate"].dt.month
            ).sum()
        )

    if "InvoiceDay" in transaction_df.columns:

        print(
            "Day Mismatches:",
            (
                transaction_df["InvoiceDay"]
                != transaction_df["InvoiceDate"].dt.day
            ).sum()
        )

    if "MonthName" in transaction_df.columns:

        print(
            "Month Name Mismatches:",
            (
                transaction_df["MonthName"]
                != transaction_df["InvoiceDate"].dt.month_name()
            ).sum()
        )

    if "DayOfWeek" in transaction_df.columns:

        print(
            "Day Name Mismatches:",
            (
                transaction_df["DayOfWeek"]
                != transaction_df["InvoiceDate"].dt.day_name()
            ).sum()
        )


# ------------------------------------------------------------
# 3.14 Validate Revenue Feature
# ------------------------------------------------------------

required_revenue_cols = {
    "Quantity",
    "UnitPrice",
    "Revenue"
}

if required_revenue_cols.issubset(
    transaction_df.columns
):

    revenue_check = np.isclose(
        transaction_df["Revenue"],
        transaction_df["Quantity"]
        * transaction_df["UnitPrice"],
        equal_nan=True
    )

    print(
        "Incorrect Revenue Records:",
        (~revenue_check).sum()
    )


# ------------------------------------------------------------
# 3.15 Validate Customer Features
# ------------------------------------------------------------

required_aov_cols = {
    "TotalRevenue",
    "OrderCount",
    "AverageOrderValue"
}

if required_aov_cols.issubset(
    customer_df.columns
):

    valid_orders = (
        customer_df["OrderCount"] != 0
    )

    aov_check = np.isclose(
        customer_df.loc[
            valid_orders,
            "AverageOrderValue"
        ],

        (
            customer_df.loc[
                valid_orders,
                "TotalRevenue"
            ]
            /
            customer_df.loc[
                valid_orders,
                "OrderCount"
            ]
        ),

        equal_nan=True
    )

    print(
        "Incorrect Average Order Value Records:",
        (~aov_check).sum()
    )


# ------------------------------------------------------------
# 3.16 Validate Product Features
# ------------------------------------------------------------

for column in [
    "TotalRevenue",
    "TotalQuantity",
    "TransactionCount"
]:

    if column in product_df.columns:

        print(
            f"{column} < 0:",
            (product_df[column] < 0).sum()
        )


if "TransactionCount" in product_df.columns:

    print(
        "TransactionCount <= 0:",
        (
            product_df["TransactionCount"] <= 0
        ).sum()
    )


# ------------------------------------------------------------
# 3.17 Categorical Feature Validation
# ------------------------------------------------------------

for name, df in datasets.items():

    categorical_cols = (
        df.select_dtypes(
            include=["object", "category"]
        )
        .columns
        .tolist()
    )

    print(f"\n{name} Categorical Features:")
    print(categorical_cols)


if "Country" in transaction_df.columns:

    print("\nCountry Distribution:")

    display(
        transaction_df["Country"]
        .value_counts()
        .to_frame("Count")
    )


if "DayOfWeek" in transaction_df.columns:

    print("\nDay-of-Week Distribution:")

    display(
        transaction_df["DayOfWeek"]
        .value_counts()
        .to_frame("Count")
    )


# ------------------------------------------------------------
# 3.18 Final Validation Summary
# ------------------------------------------------------------

validation_summary = pd.DataFrame({

    "Dataset": list(datasets.keys()),

    "Rows": [
        len(df)
        for df in datasets.values()
    ],

    "Columns": [
        df.shape[1]
        for df in datasets.values()
    ],

    "Missing_Values": [
        df.isna().sum().sum()
        for df in datasets.values()
    ],

    "Exact_Duplicates": [
        df.duplicated().sum()
        for df in datasets.values()
    ]
})

print("\nEDA DATASET VALIDATION SUMMARY")
display(validation_summary)


# ============================================================
# 4. UNIVARIATE ANALYSIS
# ============================================================


# ------------------------------------------------------------
# 4.1 Identify Numerical and Categorical Features
# ------------------------------------------------------------

for name, df in datasets.items():

    numerical_cols = (
        df.select_dtypes(include=np.number)
        .columns
        .tolist()
    )

    categorical_cols = (
        df.select_dtypes(
            include=["object", "category"]
        )
        .columns
        .tolist()
    )

    print(f"\n{name} Dataset")
    print("-" * 60)

    print("Numerical Features:")
    print(numerical_cols)

    print("\nCategorical Features:")
    print(categorical_cols)


# ------------------------------------------------------------
# 4.2 Select Meaningful Quantitative Features
# ------------------------------------------------------------

# These are analytical measurements rather than identifiers,
# flags or calendar labels.

transaction_analysis_features = [
    col for col in [
        "Quantity",
        "UnitPrice",
        "Revenue"
    ]
    if col in transaction_df.columns
]


product_analysis_features = [
    col for col in [
        "TotalRevenue",
        "TotalQuantity",
        "TransactionCount",
        "AverageUnitPrice",
        "AverageRevenue",
        "CustomerCount"
    ]
    if col in product_df.columns
]


customer_analysis_features = [
    col for col in [
        "TotalRevenue",
        "OrderCount",
        "AverageOrderValue",
        "TotalQuantity",
        "UniqueProducts",
        "TransactionCount"
    ]
    if col in customer_df.columns
]


print(
    "Transaction Analysis Features:",
    transaction_analysis_features
)

print(
    "Product Analysis Features:",
    product_analysis_features
)

print(
    "Customer Analysis Features:",
    customer_analysis_features
)


# ------------------------------------------------------------
# 4.3 Extended Numerical Summary Statistics
# ------------------------------------------------------------

def numerical_summary(df, columns):

    if len(columns) == 0:
        return pd.DataFrame()

    summary = df[columns].describe().T

    # Central tendency
    summary["median"] = (
        df[columns].median()
    )

    # Dispersion
    summary["variance"] = (
        df[columns].var()
    )

    summary["IQR"] = (
        df[columns].quantile(0.75)
        -
        df[columns].quantile(0.25)
    )

    summary["range"] = (
        df[columns].max()
        -
        df[columns].min()
    )

    # Distribution shape
    summary["skewness"] = (
        df[columns].skew()
    )

    summary["kurtosis"] = (
        df[columns].kurt()
    )

    # Relative variability
    summary["CV"] = (
        summary["std"]
        /
        summary["mean"].abs()
    )

    return summary.round(2)


print("\nTRANSACTION-LEVEL NUMERICAL SUMMARY")

transaction_summary = numerical_summary(
    transaction_df,
    transaction_analysis_features
)

display(transaction_summary)


print("\nPRODUCT-LEVEL NUMERICAL SUMMARY")

product_summary = numerical_summary(
    product_df,
    product_analysis_features
)

display(product_summary)


print("\nCUSTOMER-LEVEL NUMERICAL SUMMARY")

customer_summary = numerical_summary(
    customer_df,
    customer_analysis_features
)

display(customer_summary)


# ------------------------------------------------------------
# 4.4 Percentile Analysis
# ------------------------------------------------------------

def percentile_summary(df, columns):

    if len(columns) == 0:
        return pd.DataFrame()

    percentiles = [
        0.01,
        0.05,
        0.25,
        0.50,
        0.75,
        0.95,
        0.99
    ]

    summary = (
        df[columns]
        .quantile(percentiles)
        .T
    )

    summary.columns = [
        "P01",
        "P05",
        "P25",
        "P50",
        "P75",
        "P95",
        "P99"
    ]

    return summary.round(2)


print("\nTRANSACTION PERCENTILES")
display(
    percentile_summary(
        transaction_df,
        transaction_analysis_features
    )
)

print("\nPRODUCT PERCENTILES")
display(
    percentile_summary(
        product_df,
        product_analysis_features
    )
)

print("\nCUSTOMER PERCENTILES")
display(
    percentile_summary(
        customer_df,
        customer_analysis_features
    )
)


# ------------------------------------------------------------
# 4.5 Analytical Feature Quality Summary
# ------------------------------------------------------------

def feature_quality_summary(df, columns):

    if len(columns) == 0:
        return pd.DataFrame()

    return pd.DataFrame({

        "Missing":
            df[columns].isna().sum(),

        "Infinite":
            np.isinf(df[columns]).sum(),

        "Zero":
            (df[columns] == 0).sum(),

        "Negative":
            (df[columns] < 0).sum()

    })


print("\nTRANSACTION FEATURE QUALITY")
display(
    feature_quality_summary(
        transaction_df,
        transaction_analysis_features
    )
)

print("\nPRODUCT FEATURE QUALITY")
display(
    feature_quality_summary(
        product_df,
        product_analysis_features
    )
)

print("\nCUSTOMER FEATURE QUALITY")
display(
    feature_quality_summary(
        customer_df,
        customer_analysis_features
    )
)


# ------------------------------------------------------------
# 4.6 Histogram Analysis
# ------------------------------------------------------------

def plot_histograms(df, columns, dataset_name):

    for column in columns:

        fig, ax = plt.subplots(
            figsize=(10, 5)
        )

        sns.histplot(
            data=df,
            x=column,
            bins=40,
            ax=ax
        )

        ax.set_title(
            f"{dataset_name}: Distribution of {column}"
        )

        ax.set_xlabel(column)
        ax.set_ylabel("Frequency")

        plt.tight_layout()
        plt.show()


plot_histograms(
    transaction_df,
    transaction_analysis_features,
    "Transaction"
)

plot_histograms(
    product_df,
    product_analysis_features,
    "Product"
)

plot_histograms(
    customer_df,
    customer_analysis_features,
    "Customer"
)


# ------------------------------------------------------------
# 4.7 KDE Analysis
# ------------------------------------------------------------

def plot_kde(df, columns, dataset_name):

    for column in columns:

        values = (
            df[column]
            .replace([np.inf, -np.inf], np.nan)
            .dropna()
        )

        if values.nunique() <= 1:
            continue

        fig, ax = plt.subplots(
            figsize=(10, 5)
        )

        sns.kdeplot(
            x=values,
            fill=True,
            ax=ax
        )

        ax.set_title(
            f"{dataset_name}: KDE of {column}"
        )

        ax.set_xlabel(column)
        ax.set_ylabel("Density")

        plt.tight_layout()
        plt.show()


plot_kde(
    transaction_df,
    transaction_analysis_features,
    "Transaction"
)

plot_kde(
    product_df,
    product_analysis_features,
    "Product"
)

plot_kde(
    customer_df,
    customer_analysis_features,
    "Customer"
)


# ------------------------------------------------------------
# 4.8 ECDF Analysis
# ------------------------------------------------------------

def plot_ecdf(df, columns, dataset_name):

    for column in columns:

        fig, ax = plt.subplots(
            figsize=(10, 5)
        )

        sns.ecdfplot(
            data=df,
            x=column,
            ax=ax
        )

        ax.set_title(
            f"{dataset_name}: ECDF of {column}"
        )

        ax.set_xlabel(column)
        ax.set_ylabel(
            "Cumulative Proportion"
        )

        plt.tight_layout()
        plt.show()


plot_ecdf(
    transaction_df,
    transaction_analysis_features,
    "Transaction"
)

plot_ecdf(
    product_df,
    product_analysis_features,
    "Product"
)

plot_ecdf(
    customer_df,
    customer_analysis_features,
    "Customer"
)


# ------------------------------------------------------------
# 4.9 Boxplot Analysis
# ------------------------------------------------------------

def plot_boxplots(df, columns, dataset_name):

    for column in columns:

        fig, ax = plt.subplots(
            figsize=(10, 4)
        )

        sns.boxplot(
            data=df,
            x=column,
            ax=ax
        )

        ax.set_title(
            f"{dataset_name}: Boxplot of {column}"
        )

        ax.set_xlabel(column)

        plt.tight_layout()
        plt.show()


plot_boxplots(
    transaction_df,
    transaction_analysis_features,
    "Transaction"
)

plot_boxplots(
    product_df,
    product_analysis_features,
    "Product"
)

plot_boxplots(
    customer_df,
    customer_analysis_features,
    "Customer"
)


# ------------------------------------------------------------
# 4.10 Categorical Frequency Analysis
# ------------------------------------------------------------

def categorical_summary(df, column):

    count = df[column].value_counts(
        dropna=False
    )

    percentage = (
        df[column]
        .value_counts(
            normalize=True,
            dropna=False
        )
        * 100
    )

    return pd.DataFrame({
        "Count": count,
        "Percentage": percentage.round(2)
    })


# Country
if "Country" in transaction_df.columns:

    print("\nCOUNTRY FREQUENCY SUMMARY")

    display(
        categorical_summary(
            transaction_df,
            "Country"
        )
    )


# Transaction Status
if "TransactionStatus" in transaction_df.columns:

    print("\nTRANSACTION STATUS SUMMARY")

    display(
        categorical_summary(
            transaction_df,
            "TransactionStatus"
        )
    )


# Day of Week
if "DayOfWeek" in transaction_df.columns:

    print("\nDAY-OF-WEEK SUMMARY")

    display(
        categorical_summary(
            transaction_df,
            "DayOfWeek"
        )
    )


# Month
if "MonthName" in transaction_df.columns:

    print("\nMONTH SUMMARY")

    display(
        categorical_summary(
            transaction_df,
            "MonthName"
        )
    )


# ------------------------------------------------------------
# 4.11 Categorical Visualisation
# ------------------------------------------------------------

# Top countries by transaction-line count

if "Country" in transaction_df.columns:

    top_countries = (
        transaction_df["Country"]
        .value_counts()
        .head(15)
        .sort_values()
    )

    fig, ax = plt.subplots(
        figsize=(10, 7)
    )

    ax.barh(
        top_countries.index,
        top_countries.values
    )

    ax.set_title(
        "Top 15 Countries by Transaction-Line Count"
    )

    ax.set_xlabel(
        "Transaction-Line Count"
    )

    ax.set_ylabel("Country")

    plt.tight_layout()
    plt.show()


# Transaction status

if "TransactionStatus" in transaction_df.columns:

    fig, ax = plt.subplots(
        figsize=(8, 5)
    )

    sns.countplot(
        data=transaction_df,
        x="TransactionStatus",
        ax=ax
    )

    ax.set_title(
        "Transaction Status Distribution"
    )

    ax.set_xlabel(
        "Transaction Status"
    )

    ax.set_ylabel("Count")

    plt.xticks(rotation=30)

    plt.tight_layout()
    plt.show()


# Day-of-week ordering

weekday_order = [
    "Monday",
    "Tuesday",
    "Wednesday",
    "Thursday",
    "Friday",
    "Saturday",
    "Sunday"
]

if "DayOfWeek" in transaction_df.columns:

    fig, ax = plt.subplots(
        figsize=(10, 5)
    )

    sns.countplot(
        data=transaction_df,
        x="DayOfWeek",
        order=[
            day for day in weekday_order
            if day in
            transaction_df["DayOfWeek"].unique()
        ],
        ax=ax
    )

    ax.set_title(
        "Transaction Activity by Day of Week"
    )

    ax.set_xlabel("Day of Week")
    ax.set_ylabel(
        "Transaction-Line Count"
    )

    plt.xticks(rotation=30)

    plt.tight_layout()
    plt.show()


# ============================================================
# 5. BIVARIATE ANALYSIS
# ============================================================


# ------------------------------------------------------------
# 5.1 Numerical vs Numerical Analysis
# ------------------------------------------------------------

def scatter_analysis(
    df,
    x,
    y,
    title,
    sample_size=None
):

    plot_df = df[[x, y]].dropna()

    # Sampling prevents severe overplotting
    if (
        sample_size is not None
        and len(plot_df) > sample_size
    ):

        plot_df = plot_df.sample(
            sample_size,
            random_state=42
        )

    fig, ax = plt.subplots(
        figsize=(9, 6)
    )

    sns.scatterplot(
        data=plot_df,
        x=x,
        y=y,
        alpha=0.5,
        ax=ax
    )

    ax.set_title(title)

    plt.tight_layout()
    plt.show()


if {
    "Quantity",
    "Revenue"
}.issubset(transaction_df.columns):

    scatter_analysis(
        transaction_df,
        "Quantity",
        "Revenue",
        "Quantity vs Revenue",
        sample_size=20000
    )


if {
    "UnitPrice",
    "Revenue"
}.issubset(transaction_df.columns):

    scatter_analysis(
        transaction_df,
        "UnitPrice",
        "Revenue",
        "Unit Price vs Revenue",
        sample_size=20000
    )


if {
    "OrderCount",
    "TotalRevenue"
}.issubset(customer_df.columns):

    scatter_analysis(
        customer_df,
        "OrderCount",
        "TotalRevenue",
        "Customer Order Count vs Total Revenue"
    )


if {
    "AverageOrderValue",
    "TotalRevenue"
}.issubset(customer_df.columns):

    scatter_analysis(
        customer_df,
        "AverageOrderValue",
        "TotalRevenue",
        "Average Order Value vs Total Revenue"
    )


# ------------------------------------------------------------
# 5.2 Categorical vs Numerical Analysis
# ------------------------------------------------------------

# Revenue by transaction status

if {
    "TransactionStatus",
    "Revenue"
}.issubset(transaction_df.columns):

    fig, ax = plt.subplots(
        figsize=(9, 5)
    )

    sns.boxplot(
        data=transaction_df,
        x="TransactionStatus",
        y="Revenue",
        showfliers=False,
        ax=ax
    )

    ax.set_title(
        "Revenue Distribution by Transaction Status"
    )

    ax.set_xlabel(
        "Transaction Status"
    )

    ax.set_ylabel("Revenue")

    plt.xticks(rotation=30)

    plt.tight_layout()
    plt.show()


# Revenue by weekday

if {
    "DayOfWeek",
    "Revenue"
}.issubset(transaction_df.columns):

    weekday_revenue = (
        transaction_df
        .groupby(
            "DayOfWeek",
            observed=True
        )["Revenue"]
        .sum()
        .reindex(weekday_order)
        .dropna()
        .reset_index()
    )

    fig, ax = plt.subplots(
        figsize=(10, 5)
    )

    sns.barplot(
        data=weekday_revenue,
        x="DayOfWeek",
        y="Revenue",
        ax=ax
    )

    ax.set_title(
        "Total Revenue by Day of Week"
    )

    ax.set_xlabel("Day of Week")
    ax.set_ylabel("Total Revenue")

    plt.xticks(rotation=30)

    plt.tight_layout()
    plt.show()


# ------------------------------------------------------------
# 5.3 Country vs Revenue
# ------------------------------------------------------------

if {
    "Country",
    "Revenue"
}.issubset(transaction_df.columns):

    country_revenue = (
        transaction_df
        .groupby(
            "Country",
            observed=True
        )["Revenue"]
        .sum()
        .sort_values(
            ascending=False
        )
        .head(15)
        .sort_values()
    )

    fig, ax = plt.subplots(
        figsize=(10, 7)
    )

    ax.barh(
        country_revenue.index,
        country_revenue.values
    )

    ax.set_title(
        "Top 15 Countries by Total Revenue"
    )

    ax.set_xlabel("Total Revenue")
    ax.set_ylabel("Country")

    plt.tight_layout()
    plt.show()


# ============================================================
# 6. MULTIVARIATE ANALYSIS
# ============================================================


# ------------------------------------------------------------
# 6.1 Multivariate Customer Scatterplot
# ------------------------------------------------------------

if {
    "OrderCount",
    "TotalRevenue",
    "AverageOrderValue"
}.issubset(customer_df.columns):

    fig, ax = plt.subplots(
        figsize=(10, 6)
    )

    sns.scatterplot(
        data=customer_df,
        x="OrderCount",
        y="TotalRevenue",
        size="AverageOrderValue",
        alpha=0.6,
        ax=ax
    )

    ax.set_title(
        "Customer Revenue, Order Frequency and Average Order Value"
    )

    plt.tight_layout()
    plt.show()


# ------------------------------------------------------------
# 6.2 Pairwise Customer Relationships
# ------------------------------------------------------------

pairplot_features = [
    col for col in [
        "TotalRevenue",
        "OrderCount",
        "AverageOrderValue",
        "TotalQuantity",
        "UniqueProducts"
    ]
    if col in customer_df.columns
]


if len(pairplot_features) >= 2:

    pairplot_df = (
        customer_df[
            pairplot_features
        ]
        .replace(
            [np.inf, -np.inf],
            np.nan
        )
        .dropna()
    )

    # Sampling keeps pairplot efficient
    if len(pairplot_df) > 3000:

        pairplot_df = pairplot_df.sample(
            3000,
            random_state=42
        )

    sns.pairplot(
        pairplot_df,
        corner=True
    )

    plt.show()


# ============================================================
# 7. CORRELATION ANALYSIS
# ============================================================


# ------------------------------------------------------------
# 7.1 Pearson Correlation
# ------------------------------------------------------------

def correlation_heatmap(
    df,
    columns,
    method,
    title
):

    valid_columns = [
        col for col in columns
        if col in df.columns
    ]

    if len(valid_columns) < 2:
        return

    correlation_matrix = (
        df[valid_columns]
        .corr(method=method)
    )

    mask = np.triu(
        np.ones_like(
            correlation_matrix,
            dtype=bool
        )
    )

    fig, ax = plt.subplots(
        figsize=(10, 7)
    )

    sns.heatmap(
        correlation_matrix,
        mask=mask,
        annot=True,
        fmt=".2f",
        center=0,
        ax=ax
    )

    ax.set_title(title)

    plt.tight_layout()
    plt.show()


correlation_heatmap(
    transaction_df,
    transaction_analysis_features,
    "pearson",
    "Transaction-Level Pearson Correlation"
)

correlation_heatmap(
    product_df,
    product_analysis_features,
    "pearson",
    "Product-Level Pearson Correlation"
)

correlation_heatmap(
    customer_df,
    customer_analysis_features,
    "pearson",
    "Customer-Level Pearson Correlation"
)


# ------------------------------------------------------------
# 7.2 Spearman Correlation
# ------------------------------------------------------------

correlation_heatmap(
    transaction_df,
    transaction_analysis_features,
    "spearman",
    "Transaction-Level Spearman Correlation"
)

correlation_heatmap(
    product_df,
    product_analysis_features,
    "spearman",
    "Product-Level Spearman Correlation"
)

correlation_heatmap(
    customer_df,
    customer_analysis_features,
    "spearman",
    "Customer-Level Spearman Correlation"
)


# ============================================================
# 8. OUTLIER AND DISTRIBUTION ANALYSIS
# ============================================================


# ------------------------------------------------------------
# 8.1 IQR Outlier Summary
# ------------------------------------------------------------

def iqr_outlier_summary(df, columns):

    results = []

    for column in columns:

        series = (
            df[column]
            .replace(
                [np.inf, -np.inf],
                np.nan
            )
            .dropna()
        )

        q1 = series.quantile(0.25)
        q3 = series.quantile(0.75)

        iqr = q3 - q1

        lower_bound = (
            q1 - 1.5 * iqr
        )

        upper_bound = (
            q3 + 1.5 * iqr
        )

        outlier_mask = (
            (series < lower_bound)
            |
            (series > upper_bound)
        )

        outlier_count = (
            outlier_mask.sum()
        )

        outlier_percentage = (
            outlier_count
            /
            len(series)
            * 100
        )

        results.append({

            "Feature":
                column,

            "Q1":
                q1,

            "Q3":
                q3,

            "IQR":
                iqr,

            "Lower_Bound":
                lower_bound,

            "Upper_Bound":
                upper_bound,

            "Outlier_Count":
                outlier_count,

            "Outlier_Percentage":
                outlier_percentage
        })

    return (
        pd.DataFrame(results)
        .round(2)
    )


print("\nTRANSACTION IQR OUTLIER SUMMARY")

transaction_outliers = iqr_outlier_summary(
    transaction_df,
    transaction_analysis_features
)

display(transaction_outliers)


print("\nPRODUCT IQR OUTLIER SUMMARY")

product_outliers = iqr_outlier_summary(
    product_df,
    product_analysis_features
)

display(product_outliers)


print("\nCUSTOMER IQR OUTLIER SUMMARY")

customer_outliers = iqr_outlier_summary(
    customer_df,
    customer_analysis_features
)

display(customer_outliers)


# ------------------------------------------------------------
# 8.2 Log-Scale Distribution Investigation
# ------------------------------------------------------------

# Log visualisation is useful for strongly right-skewed,
# strictly positive variables. It does NOT modify the dataset.

def log_distribution_plot(
    df,
    column,
    dataset_name
):

    positive_values = (
        df.loc[
            df[column] > 0,
            column
        ]
        .replace(
            [np.inf, -np.inf],
            np.nan
        )
        .dropna()
    )

    if positive_values.empty:
        return

    log_values = np.log1p(
        positive_values
    )

    fig, ax = plt.subplots(
        figsize=(10, 5)
    )

    sns.histplot(
        x=log_values,
        bins=40,
        kde=True,
        ax=ax
    )

    ax.set_title(
        f"{dataset_name}: Log-Transformed Distribution of {column}"
    )

    ax.set_xlabel(
        f"log(1 + {column})"
    )

    ax.set_ylabel("Frequency")

    plt.tight_layout()
    plt.show()


# Only investigate strongly skewed positive features

for column in transaction_analysis_features:

    if (
        column in transaction_df.columns
        and
        transaction_df[column]
        .dropna()
        .skew() > 1
    ):

        log_distribution_plot(
            transaction_df,
            column,
            "Transaction"
        )


for column in product_analysis_features:

    if (
        column in product_df.columns
        and
        product_df[column]
        .dropna()
        .skew() > 1
    ):

        log_distribution_plot(
            product_df,
            column,
            "Product"
        )


for column in customer_analysis_features:

    if (
        column in customer_df.columns
        and
        customer_df[column]
        .dropna()
        .skew() > 1
    ):

        log_distribution_plot(
            customer_df,
            column,
            "Customer"
        )


# ============================================================
# 9. TEMPORAL EXPLORATORY ANALYSIS
# ============================================================


# ------------------------------------------------------------
# 9.1 Monthly Revenue
# ------------------------------------------------------------

if {
    "InvoiceDate",
    "Revenue"
}.issubset(transaction_df.columns):

    monthly_revenue = (
        transaction_df
        .set_index("InvoiceDate")
        ["Revenue"]
        .resample("ME")
        .sum()
    )

    fig, ax = plt.subplots(
        figsize=(12, 6)
    )

    ax.plot(
        monthly_revenue.index,
        monthly_revenue.values,
        marker="o"
    )

    ax.set_title(
        "Monthly Revenue Trend"
    )

    ax.set_xlabel("Month")
    ax.set_ylabel("Revenue")

    plt.xticks(rotation=45)

    plt.tight_layout()
    plt.show()


# ------------------------------------------------------------
# 9.2 Daily Revenue and Rolling Average
# ------------------------------------------------------------

if {
    "InvoiceDate",
    "Revenue"
}.issubset(transaction_df.columns):

    daily_revenue = (
        transaction_df
        .set_index("InvoiceDate")
        ["Revenue"]
        .resample("D")
        .sum()
    )

    rolling_7d = (
        daily_revenue
        .rolling(7)
        .mean()
    )

    fig, ax = plt.subplots(
        figsize=(13, 6)
    )

    ax.plot(
        daily_revenue.index,
        daily_revenue.values,
        label="Daily Revenue",
        alpha=0.5
    )

    ax.plot(
        rolling_7d.index,
        rolling_7d.values,
        label="7-Day Rolling Average"
    )

    ax.set_title(
        "Daily Revenue with 7-Day Rolling Average"
    )

    ax.set_xlabel("Date")
    ax.set_ylabel("Revenue")

    ax.legend()

    plt.tight_layout()
    plt.show()


# ------------------------------------------------------------
# 9.3 Month × Revenue
# ------------------------------------------------------------

if {
    "InvoiceMonth",
    "Revenue"
}.issubset(transaction_df.columns):

    monthly_number_revenue = (
        transaction_df
        .groupby(
            "InvoiceMonth",
            observed=True
        )["Revenue"]
        .sum()
        .reset_index()
    )

    fig, ax = plt.subplots(
        figsize=(10, 5)
    )

    sns.lineplot(
        data=monthly_number_revenue,
        x="InvoiceMonth",
        y="Revenue",
        marker="o",
        ax=ax
    )

    ax.set_title(
        "Revenue by Calendar Month"
    )

    ax.set_xlabel("Month")
    ax.set_ylabel("Revenue")

    ax.set_xticks(
        sorted(
            monthly_number_revenue[
                "InvoiceMonth"
            ].unique()
        )
    )

    plt.tight_layout()
    plt.show()


# ------------------------------------------------------------
# 9.4 Day × Hour Heatmap
# ------------------------------------------------------------

hour_column = None

for possible_column in [
    "InvoiceHour",
    "Hour"
]:

    if possible_column in transaction_df.columns:

        hour_column = possible_column
        break


if (
    hour_column is not None
    and
    "DayOfWeek" in transaction_df.columns
    and
    "Revenue" in transaction_df.columns
):

    day_hour_revenue = (
        transaction_df
        .pivot_table(
            values="Revenue",
            index="DayOfWeek",
            columns=hour_column,
            aggfunc="sum",
            fill_value=0
        )
        .reindex(
            [
                day for day
                in weekday_order
                if day in
                transaction_df[
                    "DayOfWeek"
                ].unique()
            ]
        )
    )

    fig, ax = plt.subplots(
        figsize=(13, 6)
    )

    sns.heatmap(
        day_hour_revenue,
        ax=ax
    )

    ax.set_title(
        "Revenue Heatmap: Day of Week × Hour"
    )

    ax.set_xlabel("Hour")
    ax.set_ylabel("Day of Week")

    plt.tight_layout()
    plt.show()


# ============================================================
# 10. PRODUCT-LEVEL EXPLORATORY ANALYSIS
# ============================================================


# ------------------------------------------------------------
# 10.1 Top Products by Revenue
# ------------------------------------------------------------

if {
    "StockCode",
    "TotalRevenue"
}.issubset(product_df.columns):

    top_products_revenue = (
        product_df
        .nlargest(
            15,
            "TotalRevenue"
        )
        .sort_values(
            "TotalRevenue"
        )
    )

    fig, ax = plt.subplots(
        figsize=(10, 7)
    )

    ax.barh(
        top_products_revenue["StockCode"]
        .astype(str),

        top_products_revenue[
            "TotalRevenue"
        ]
    )

    ax.set_title(
        "Top 15 Products by Revenue"
    )

    ax.set_xlabel(
        "Total Revenue"
    )

    ax.set_ylabel("Stock Code")

    plt.tight_layout()
    plt.show()


# ------------------------------------------------------------
# 10.2 Top Products by Quantity
# ------------------------------------------------------------

if {
    "StockCode",
    "TotalQuantity"
}.issubset(product_df.columns):

    top_products_quantity = (
        product_df
        .nlargest(
            15,
            "TotalQuantity"
        )
        .sort_values(
            "TotalQuantity"
        )
    )

    fig, ax = plt.subplots(
        figsize=(10, 7)
    )

    ax.barh(
        top_products_quantity[
            "StockCode"
        ].astype(str),

        top_products_quantity[
            "TotalQuantity"
        ]
    )

    ax.set_title(
        "Top 15 Products by Quantity"
    )

    ax.set_xlabel(
        "Total Quantity"
    )

    ax.set_ylabel("Stock Code")

    plt.tight_layout()
    plt.show()


# ============================================================
# 11. CUSTOMER-LEVEL EXPLORATORY ANALYSIS
# ============================================================


# ------------------------------------------------------------
# 11.1 Top Customers by Revenue
# ------------------------------------------------------------

if {
    "CustomerID",
    "TotalRevenue"
}.issubset(customer_df.columns):

    top_customers = (
        customer_df
        .nlargest(
            15,
            "TotalRevenue"
        )
        .sort_values(
            "TotalRevenue"
        )
    )

    fig, ax = plt.subplots(
        figsize=(10, 7)
    )

    ax.barh(
        top_customers[
            "CustomerID"
        ].astype(str),

        top_customers[
            "TotalRevenue"
        ]
    )

    ax.set_title(
        "Top 15 Customers by Revenue"
    )

    ax.set_xlabel(
        "Total Revenue"
    )

    ax.set_ylabel(
        "Customer ID"
    )

    plt.tight_layout()
    plt.show()


# ============================================================
# 12. FINAL EDA SUMMARY TABLES
# ============================================================


# ------------------------------------------------------------
# 12.1 Dataset-Level Summary
# ------------------------------------------------------------

eda_summary = pd.DataFrame({

    "Dataset": [
        "Transaction",
        "Product",
        "Customer"
    ],

    "Rows": [
        len(transaction_df),
        len(product_df),
        len(customer_df)
    ],

    "Columns": [
        transaction_df.shape[1],
        product_df.shape[1],
        customer_df.shape[1]
    ],

    "Numerical_Analysis_Features": [
        len(transaction_analysis_features),
        len(product_analysis_features),
        len(customer_analysis_features)
    ],

    "Missing_Values": [
        transaction_df.isna().sum().sum(),
        product_df.isna().sum().sum(),
        customer_df.isna().sum().sum()
    ],

    "Exact_Duplicates": [
        transaction_df.duplicated().sum(),
        product_df.duplicated().sum(),
        customer_df.duplicated().sum()
    ]
})

print("\nFINAL EDA DATASET SUMMARY")

display(eda_summary)


# ------------------------------------------------------------
# 12.2 Skewness Summary
# ------------------------------------------------------------

def skewness_summary(
    df,
    columns,
    dataset_name
):

    if len(columns) == 0:
        return pd.DataFrame()

    result = pd.DataFrame({

        "Dataset":
            dataset_name,

        "Feature":
            columns,

        "Skewness":
            [
                df[column].skew()
                for column in columns
            ]
    })

    result["Distribution_Comment"] = (
        result["Skewness"]
        .apply(
            lambda x:
            "Strong Right Skew"
            if x > 1
            else
            "Moderate Right Skew"
            if x > 0.5
            else
            "Strong Left Skew"
            if x < -1
            else
            "Moderate Left Skew"
            if x < -0.5
            else
            "Approximately Symmetric"
        )
    )

    return result


skewness_results = pd.concat(
    [
        skewness_summary(
            transaction_df,
            transaction_analysis_features,
            "Transaction"
        ),

        skewness_summary(
            product_df,
            product_analysis_features,
            "Product"
        ),

        skewness_summary(
            customer_df,
            customer_analysis_features,
            "Customer"
        )
    ],
    ignore_index=True
)

print("\nDISTRIBUTION SHAPE SUMMARY")

display(
    skewness_results.round(2)
)


# ============================================================
# END OF EXPLORATORY DATA ANALYSIS
# ============================================================

print(
    "\nEDA notebook completed successfully."
)
```
