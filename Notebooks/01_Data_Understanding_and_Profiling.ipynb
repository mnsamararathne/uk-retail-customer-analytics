```python
#importing the libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
#-------------------------------------------------------------------------------------------
#loading the dataset
retail_df = pd.read_excel("/Users/manesh/Documents/Manesh /Masters/Portfolios/uk-retail-customer-analytics/Data/Online Retail.xlsx")
#-------------------------------------------------------------------------------------------

#examin the dataset data strutre

#preview first 5 rows
display(retail_df.head())

print()

#preview last 5 rows
display(retail_df.tail())

print()

#examin dataset dimensions
print("dataset shape: ", retail_df.shape)
print("dataset num of rows", retail_df.shape[0])
print("dataset num of cols",retail_df.shape[1] )

print()

#examin the col names
print("\nColumn names: ")
print(retail_df.columns.tolist)

print()
#-------------------------------------------------------------------------------------------
#examin the data types and the non-null count of the dataset
print("\nDataset Information: ")
retail_df.info()

print()

#dataset datatypes
print("\nDataset datatypes: ")
print(retail_df.dtypes)

print()
#-------------------------------------------------------------------------------------------
#statistical summary of the dataset
display(retail_df.describe())

print()

#statistical summary including categorical columns
display(retail_df.describe(include="all").T)

print()
#-------------------------------------------------------------------------------------------
#check the number unique values in each column 
unique_values = retail_df.nunique().sort_values

print(unique_values)

#checking the unique values of each columns 

uniqe_description = retail_df["Description"].unique()
unique_stockcode = retail_df["StockCode"].unique()
unique_country = retail_df["Country"].unique()

print("\nUnique descriptions: ")
print(uniqe_description)

print("\nUnique stockcodes: ")
print(unique_stockcode)

print("\nunique country: ")
print(unique_country)

#check unique values with their frequencies 
unique_description_freq = retail_df["Description"].value_counts()
unique_stockcode_freq = retail_df["StockCode"].value_counts()
unique_country_freq = retail_df["Country"].unique()

print("\nUnique descriptions frequencies:")
print(unique_description_freq)

print("\nUnique stockcodes frequencies:")
print(unique_stockcode_freq)

print("\nUnique Country frequencies: ")
print(unique_country_freq)

#check propotion of unique value frequencies 
desc_freq_prop = retail_df["Description"].value_counts(normalize= True)
stockcode_freq_prop = retail_df["StockCode"].value_counts(normalize= True)
country_freq_prop = retail_df["Country"].value_counts(normalize= True)

print("\nUnique descriptions frequencies proportions:")
print(desc_freq_prop)

print("\nUnique stockcode frequencies proportions:")
print(stockcode_freq_prop)

print("\nUnique country frequencies proportions:")
print(country_freq_prop)

#Cardinality summary of dataset
cardinality_summary = pd.DataFrame(
    {
        "Data Type": retail_df.dtypes,
        "Unique Values" : retail_df.nunique(),
        "Missing Values" : retail_df.isnull().sum()

    }
) 

cardinality_summary = cardinality_summary.sort_values(
    by = "Unique Values"
)

print("\nCardinality Summary: ")
display(cardinality_summary)

#-------------------------------------------------------------------------------------------
#Missing values Identification =================================================

#checking missing values per column
missing_values_per_col = retail_df.isnull().sum()

print("\nMissing Values Per Columns: ")
print(missing_values_per_col)

#missing value percentage per column
miss_val_per = retail_df.isnull().sum() / len(retail_df) *100

print("\nMissing Value Percentage per column: ")
print(miss_val_per)

#missing value summary 
missing_summary = pd.DataFrame(
    {
        "Missing_Count" : retail_df.isnull().sum(),
        "Missing_Percentage" : (
            miss_val_per
        )
    }
)

#show the columns containing missing values 
missing_summary = missing_summary [
    missing_summary["Missing_Count"] > 0
]

#Sort highest to Lowest 
missing_summary = missing_summary.sort_values(
    by="Missing_Percentage", 
    ascending = False
)

print("\nMisisng Value Summary: ")
display(missing_summary)

#Check the total missing value count of the dataset 
total_missing = retail_df.isnull().sum().sum()
print("Total Missing Values are : ", total_missing)

#final check on dataset contains any missing value 
Any_miss_val_check = retail_df.isnull().values.any()
print("Any missing values: ", Any_miss_val_check)

#count the rows contains at least one missing value
rows_with_missing = retail_df.isnull().any(axis=1).sum()
print("Rows with missing: ", rows_with_missing)

#handling missing values -----------------------------------
# Fill missing Description values
retail_df["Description"].fillna("Unknown")

#remove rows with missing customer IDs
retail_df = retail_df.dropna(subset=["CustomerID"])

#reset dataset index after dropping rows 
retail_df = retail_df.reset_index(drop=True)

#Validation of the missing values handling
print("\nMissing Values After Handling: ")
print(retail_df.isnull().sum())


print("\nDataset shape after handling missing values: ")
print(retail_df.shape)

#Handling duplicate values -----------------------------------

#Count duplicated rows in dataset

duplicate_rows_count = retail_df.duplicated().sum()
print("\nNumber of duplicated rows in the dataset: ", duplicate_rows_count)

#Percentage of duplicated rows in the datset 
duplicate_percentage = (
    duplicate_rows_count / len(retail_df)) * 100
print(
    "\nPercentage of duplicated rows in the dataset: ",
       round(duplicate_percentage, 2),
       "%"
       )

#inspect all the copies of duplicated rows in the dataset 
duplicated_rows = retail_df[
    retail_df.duplicated(keep=False)
]

display(duplicated_rows.sort_values(
    by = ["InvoiceNo", "StockCode", "CustomerID"]

).head(50))

#figure out the exact transactions duplicated in the dataset 
transaction_duplicates = retail_df[
    retail_df.duplicated(
        subset = [
            "InvoiceNo",
            "StockCode", 
            "CustomerID"
        ], 
        keep=False
    )
]

display(
    transaction_duplicates.sort_values(
        ["InvoiceNo", "StockCode", "CustomerID"]
    ).head(50)
)

#remove exact duplicated transactions from the dataset 
rows_before = len(retail_df)

retail_df = (
    retail_df.drop_duplicates(keep="first")
    .reset_index(drop=True)
)

rows_after = len(retail_df)

print("Rows Removed:", rows_before - rows_after)
print("Remaining duplicates: ", retail_df. duplicated().sum())

#Business Rule Validation of the dataset -----------------------------------

#1. Quantity Validation 

print("Minimum Quantity: ", retail_df["Quantity"].min())
print("Maximum Quantity: ", retail_df["Quantity"].max())

print("Negative Quantity", 
    (retail_df["Quantity"] < 0).sum()
    )

print("Negative Quantity", 
    (retail_df["Quantity"] == 0).sum()
    )

print("Possitive Quantity", 
    (retail_df["Quantity"] > 0).sum()
    )

#inspect problematic/special cases

quality_issues = retail_df[
    retail_df["Quantity"] <= 0
]

print("\nQuality Issues: ")
display(quality_issues.head(20))

#Unit Price Validation 
print("Minimum Unit Price: ", retail_df["UnitPrice"].min())

print("Maximum Unit Price ", retail_df["UnitPrice"].max())

print("Negative Unit Price ", (retail_df["UnitPrice"] < 0).sum())

print("Zero Unit Price ", (retail_df["UnitPrice"] == 0).sum())

#Inspect the issues of the unit price column 

price_issues = retail_df [
    retail_df["UnitPrice"] <= 0 
]

display(price_issues.head(20))


# validation of Customer ID
print("Missing Customer ID: ", retail_df["CustomerID"].isnull().sum())


print("Unique Customer ID: ", retail_df["CustomerID"].nunique())

#fixing the customer ID Data type from float to integer 

retail_df["CustomerID"] = retail_df["CustomerID"].astype("Int64")

#inpect the customer ID Range

print("Minimum Customer ID: ", retail_df["CustomerID"].min())

print("Maximum Customer ID", retail_df["CustomerID"].max())

#Validate the Invoice Date column 

#Ensure the column is in datetime format

retail_df["InvoiceDate"] = pd.to_datetime(retail_df["InvoiceDate"],
                                          errors="coerce")

print(
    "Earliest Transaction Date: ", retail_df["InvoiceDate"].min()
)

print(
    "Latest Transaction Date: ", retail_df["InvoiceDate"].max()
)

print(
    "Invalid Transaction Dates: ", retail_df["InvoiceDate"].isnull().sum()
)

#Validate the Stockcode column 

print(
    "Missing StockCode: ", retail_df["StockCode"].isnull().sum()
)

print(
    "Unique StockCode: ", retail_df["StockCode"].nunique()
)

display(
    retail_df["StockCode"].value_counts().head(20)
)

#Validate the Description Column
print(
    "Missing Description: ", retail_df["Description"].isnull().sum()
)

print(
    "Unknown Description: ", (retail_df["Description"] == "Unknown").sum()
)

print(
    "Unique Description: ", retail_df["Description"].nunique()
)

#remove whitespace in the decsription column 

whitespace_issues = (
    retail_df["Description"].astype(str).str.strip().eq("")
.sum() )

print("Descriptions with whitespaces: ", whitespace_issues)

#Validate Country Column

print(
    "Missing Countries: ", 
    retail_df["Country"].isnull().sum()

)

print(
    "Countries Count: ", retail_df["Country"].nunique()
)

print(
    retail_df["Country"].value_counts().head(20)
)

# check inconsistent labels, whitespace or suspicious values

print(
    "Inconsistent Country Labels,  whitespace or suspicious values: ", 
    (retail_df["Country"].str.strip()) 
)

# business rule validation summary

business_rules = pd.DataFrame({

   "Rule": [
        "Quantity < 0",
        "Quantity = 0",
        "UnitPrice < 0",
        "UnitPrice = 0",
        "Missing CustomerID",
        "Missing StockCode",
        "Missing InvoiceDate",
        "Missing Country",
        "Unknown Description",
    ],
    "Records": [
        (retail_df["Quantity"] < 0).sum(),
        (retail_df["Quantity"] == 0).sum(),
        (retail_df["UnitPrice"] < 0).sum(),
        (retail_df["UnitPrice"] == 0).sum(),
        retail_df["CustomerID"].isnull().sum(),
        retail_df["StockCode"].isnull().sum(),
        retail_df["InvoiceDate"].isnull().sum(),
        retail_df["Country"].isnull().sum(),
        (retail_df["Description"] == "Unknown").sum(),
    ]
}

)

display(business_rules)

#Cancellation / Return Validation Analysis 

# 1. Identify cancellation transactions

cancellation_mark = (
    retail_df["InvoiceNo"]
    .astype(str)
    .str.startswith("C")
)

cancelled_transactions = retail_df[cancellation_mark]

print(
    "Cancellation Transaction Rows: ",
    len(cancelled_transactions)
)

display(cancelled_transactions.head(20))

#Calculate cancellation transaction percentage 

cancellation_percentage = (
    len(cancelled_transactions) / len(retail_df)
) * 100

print(
    "Calculation Percentage: ",
    round(cancellation_percentage, 2),
    "%"
)

#check quantities of cancellation transactions 

print(
   "Cancellation rows with negative quantities:  ", 
    (
        cancelled_transactions["Quantity"] < 0 
    ).sum()
)

print(
    "Cancellation rows with zero quantities: ",
    (
        cancelled_transactions["Quantity"] == 0 
    ).sum()
)

print(
    "Cancellation rows with positive quantities: ",
    (
        cancelled_transactions["Quantity"] > 0
    ).sum()
)

# reverse check - Does it have negative Quantity?, Does the InvoiceNo begin with C?


#identify all the negative quantity transactions
negative_quantity_transactions = retail_df[
    retail_df["Quantity"] < 0 
]

print(
    "Total negative quantity rows: ",
    len(negative_quantity_transactions)
)

#Check for negative quantities WITHOUT cancellation invoice


negative_without_cancellation = retail_df[
    (retail_df["Quantity"] < 0 )
    &
    (~cancellation_mark)
]

print(
   "Negative quantities without C invoice: ",
    len(negative_without_cancellation)
)

display(negative_without_cancellation.head(20))


#Cross check cancellation status and quantity

cancellation_validation = pd.crosstab(
    cancellation_mark,
    retail_df["Quantity"] < 0, 
    rownames= ["Cancellation Invoice"],
    colnames = ["Negative Quantity"]
)

display(cancellation_validation)

#create a transaction-status feature
retail_df["TransactionStatus"] = np.where(
    retail_df["InvoiceNo"]
    .astype(str)
    .str.startswith("C"),
    "Cancelled",
    "Completed"
)

retail_df["TransactionStatus"].value_counts()

#Initial Distribution Analysis 

#Indentify the categorical and numerical columns in dataset 

#numerical columns 
numerical_cols = retail_df.select_dtypes(
    include = ["int64", "float64", "Int64"]
).columns.tolist()

numerical_analysis_cols = [
    "Quantity",
    "UnitPrice"
]

#categorical columns 

categorical_cols = retail_df.select_dtypes(
    include = ["object", "category"]
).columns.tolist()

print("\nNumerical Columns: ")
display(numerical_cols)

print("\nCategorical Columns: ")
display(categorical_cols)


#Numerical descriptive distribution 

#numerical distribution summary

display(
    retail_df[numerical_analysis_cols]
    .describe()
    .T
)

#Check skewness of the numerical column values 

for col in numerical_analysis_cols:
    skewness = retail_df[col].skew()

print(
    f"{col}: {skewness:.2f}"
)

#Plot hostograms of the Numerical col value distribution 

print("\nNumerical columns distributions: ")
#Histograms 
for col in numerical_analysis_cols:
    plt.figure(figsize=(10,5))

    sns.histplot(
        data = retail_df,
        x = col,
        bins = 50,
        kde=True
    )

    plt.title(f"Distribution of {col} ")
    plt.xlabel(col)
    plt.ylabel("Frequency")

    plt.show()

#Boxplots 
print("\nNumerical columns distributions - Boxplot: ")

for col in numerical_analysis_cols:
    plt.Figure(figsize=(10,4))

sns.boxplot(
    x=retail_df[col]
)

plt.title(f"Boxplot of {col}")
plt.xlabel(col)

plt.show()

#Analyzing the possitive sales
completed_sales = retail_df[
    (retail_df["Quantity"] > 0) &
    (retail_df["UnitPrice"] > 0)
]

display(
    completed_sales[
        ["Quantity", "UnitPrice"]
    ].describe().T
)

#Categorical Distribution Analysis

categorical_analysis_cols = [
    "Country",
    "TransactionStatus"
]

for col in categorical_analysis_cols:
     print(f"\n{'=' * 50}")
     print(f"Distribution of {col}")
     print("=" * 50)

     print(
         retail_df[col]
         .value_counts(dropna=False)
         )
     print("\nPercentages: " )

     print(
         retail_df[col]
         .value_counts(
             normalize=True,
             dropna=False

         ).mul(100).round(2)
     )

#Country distribution

#examine the top countries 
country_distribution = (
    retail_df["Country"]
    .value_counts()
    .head(10)
)

display(country_distribution)

#visualizing the country distribution 
plt.figure(figsize=(10,6))

sns.barplot(
    x = country_distribution.values,
    y = country_distribution.index
)

plt.title("Top 10 countries by transaction records")
plt.xlabel("Number of transaction records")
plt.ylabel("Country")

plt.show()

#Transaction-status distribution

tran_status_distribution = (
    retail_df["TransactionStatus"]
    .value_counts()
)

display(tran_status_distribution)

#visualizing the Transaction-status distribution

plt.figure(figsize=(7,5))

sns.countplot(
    data=retail_df,
    x="TransactionStatus"
)

plt.title("Distribution of Transaction Status")
plt.xlabel("Transaction Status")
plt.ylabel("Number of Transaction Records")

plt.show()

#Initial temporal distribution

#Create temporary time components
retail_df["InvoiceYear"] = retail_df["InvoiceDate"].dt.year
retail_df["InvoiceMonth"] = retail_df["InvoiceDate"].dt.month
retail_df["InvoiceDay"] = retail_df["InvoiceDate"].dt.day
retail_df["InvoiceHour"] = retail_df["InvoiceDate"].dt.hour
retail_df["DayOfWeek"] = retail_df["InvoiceDate"].dt.day_name()

#Inspecting the temporary time components
print(retail_df["InvoiceYear"].value_counts().sort_index())
print(retail_df["InvoiceMonth"].value_counts().sort_index())
print(retail_df["DayOfWeek"].value_counts())
print(retail_df["InvoiceHour"].value_counts().sort_index())

#overall Data-Quality Summary
total_rows = len(retail_df)
total_columns = retail_df.shape[1]

total_missing = retail_df.isnull().sum().sum()
duplicate_rows = retail_df.duplicated().sum()

negative_quantity = (retail_df["Quantity"] < 0).sum()
zero_quantity = (retail_df["Quantity"] == 0).sum()

negative_price = (retail_df["UnitPrice"] < 0).sum()
zero_price = (retail_df["UnitPrice"] == 0).sum()

missing_customer_id = retail_df["CustomerID"].isnull().sum()
missing_stock_code = retail_df["StockCode"].isnull().sum()

cancelled_transactions_count = (
    retail_df["InvoiceNo"]
    .astype(str)
    .str.startswith("C")
    .sum()
)

negative_without_cancellation = (
    (retail_df["Quantity"] < 0)
    & (~retail_df["InvoiceNo"].astype(str).str.startswith("C"))
).sum()

# Data quality summary table
total_rows = len(retail_df)
total_columns = retail_df.shape[1]

total_missing = retail_df.isnull().sum().sum()
duplicate_rows = retail_df.duplicated().sum()

negative_quantity = (retail_df["Quantity"] < 0).sum()
zero_quantity = (retail_df["Quantity"] == 0).sum()

negative_price = (retail_df["UnitPrice"] < 0).sum()
zero_price = (retail_df["UnitPrice"] == 0).sum()

missing_customer_id = retail_df["CustomerID"].isnull().sum()
missing_stock_code = retail_df["StockCode"].isnull().sum()

cancelled_transactions = (
    retail_df["InvoiceNo"]
    .astype(str)
    .str.startswith("C")
    .sum()
)

negative_without_cancellation = (
    (retail_df["Quantity"] < 0)
    &
    (~retail_df["InvoiceNo"]
        .astype(str)
        .str.startswith("C"))
).sum()


data_quality_summary = pd.DataFrame({

    "Check": [
        "Total Rows",
        "Total Columns",
        "Total Missing Values",
        "Exact Duplicate Rows",
        "Missing Customer IDs",
        "Missing Stock Codes",
        "Negative Quantities",
        "Zero Quantities",
        "Negative Unit Prices",
        "Zero Unit Prices",
        "Cancellation Transactions",
        "Negative Quantity Without Cancellation"
    ],

    "Result": [
        total_rows,
        total_columns,
        total_missing,
        duplicate_rows,
        missing_customer_id,
        missing_stock_code,
        negative_quantity,
        zero_quantity,
        negative_price,
        zero_price,
        cancelled_transactions,
        negative_without_cancellation
    ]
})

print("\nData Quality Summary: ")
display(data_quality_summary)

#Column level data quality summary

# ============================================================
# Column-Level Data Quality Summary
# ============================================================

column_quality = pd.DataFrame({

    "Data_Type": retail_df.dtypes.astype(str),

    "Missing_Count":
        retail_df.isnull().sum(),

    "Missing_Percentage":
        (retail_df.isnull().mean() * 100).round(2),

    "Unique_Values":
        retail_df.nunique(),

    "Cardinality_Percentage":
        (
            retail_df.nunique()
            / len(retail_df)
            * 100
        ).round(2)
})

display(column_quality)
```
