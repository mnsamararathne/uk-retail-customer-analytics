```python
# ============================================================
# 06_Business_Insights_and_KPI_Analysis.ipynb
#
# Online Retail Customer Analytics
#
# Business Insights, KPI Analysis and
# Streamlit Dashboard Data Preparation
# ============================================================


# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from pathlib import Path


# ------------------------------------------------------------
# Display Settings
# ------------------------------------------------------------

pd.set_option(
    "display.max_columns",
    None
)

pd.set_option(
    "display.float_format",
    lambda x: f"{x:,.2f}"
)


print("Libraries imported successfully.")



# ============================================================
# 2. LOAD ANALYTICAL DATASETS
# ============================================================


# ------------------------------------------------------------
# 2.1 Transaction-Level Dataset
# ------------------------------------------------------------

transaction_df = pd.read_csv(
    "../Data/retail_transaction_features.csv"
)


# ------------------------------------------------------------
# 2.2 Product-Level Dataset
# ------------------------------------------------------------

product_df = pd.read_csv(
    "../Data/product_features.csv"
)


# ------------------------------------------------------------
# 2.3 Customer-Level Dataset
# ------------------------------------------------------------

customer_df = pd.read_csv(
    "../Data/customer_features.csv"
)


# ------------------------------------------------------------
# 2.4 Segmented Customer Dataset
# ------------------------------------------------------------

segment_df = pd.read_csv(
    "../Data/customer_segments.csv"
)


print("All analytical datasets loaded successfully.")


print(
    "\nTransaction Dataset Shape:",
    transaction_df.shape
)

print(
    "Product Dataset Shape:",
    product_df.shape
)

print(
    "Customer Dataset Shape:",
    customer_df.shape
)

print(
    "Segmented Customer Dataset Shape:",
    segment_df.shape
)



# ============================================================
# 3. ANALYTICAL DATASET VALIDATION
# ============================================================


# ------------------------------------------------------------
# 3.1 Dataset-Level Validation
# ------------------------------------------------------------

datasets = {

    "Transaction":
        transaction_df,

    "Product":
        product_df,

    "Customer":
        customer_df,

    "Segmented Customer":
        segment_df
}


validation_results = []


for name, df in datasets.items():

    validation_results.append({

        "Dataset":
            name,

        "Rows":
            df.shape[0],

        "Columns":
            df.shape[1],

        "MissingValues":
            df.isna().sum().sum(),

        "ExactDuplicates":
            df.duplicated().sum()
    })


validation_summary = pd.DataFrame(
    validation_results
)


print(
    "\nANALYTICAL DATASET VALIDATION"
)


display(
    validation_summary
)



# ------------------------------------------------------------
# 3.2 Display Available Transaction Columns
# ------------------------------------------------------------

print(
    "\nTRANSACTION DATASET COLUMNS"
)


for number, column in enumerate(
    transaction_df.columns,
    start=1
):

    print(
        f"{number}. {column}"
    )



# ------------------------------------------------------------
# 3.3 Validate Required Raw Transaction Columns
# ------------------------------------------------------------

# Revenue is deliberately NOT included here because it can
# be derived from Quantity × UnitPrice.

required_transaction_columns = [

    "InvoiceNo",

    "StockCode",

    "Quantity",

    "InvoiceDate",

    "UnitPrice",

    "CustomerID",

    "Country"
]


missing_transaction_columns = [

    column

    for column
    in required_transaction_columns

    if column
    not in transaction_df.columns
]


if missing_transaction_columns:

    raise ValueError(

        "Missing required transaction columns: "
        +
        ", ".join(
            missing_transaction_columns
        )
    )


else:

    print(
        "\nAll required transaction columns are available."
    )



# ------------------------------------------------------------
# 3.4 Convert InvoiceDate
# ------------------------------------------------------------

transaction_df[
    "InvoiceDate"
] = pd.to_datetime(

    transaction_df[
        "InvoiceDate"
    ],

    errors="coerce"
)


invalid_dates = (

    transaction_df[
        "InvoiceDate"
    ]
    .isna()
    .sum()
)


print(
    "\nInvalid / Missing Invoice Dates:",
    invalid_dates
)


print(
    "Analysis Start Date:",
    transaction_df[
        "InvoiceDate"
    ].min()
)


print(
    "Analysis End Date:",
    transaction_df[
        "InvoiceDate"
    ].max()
)



# ============================================================
# 4. PREPARE BUSINESS ANALYSIS DATA
# ============================================================


# ------------------------------------------------------------
# 4.1 Validate / Create Revenue Feature
# ------------------------------------------------------------

# Transaction-line revenue:
#
# Revenue = Quantity × UnitPrice
#
# Negative quantities therefore produce negative revenue,
# which preserves the financial effect of cancellations /
# returns in the complete transaction dataset.


if "Revenue" not in transaction_df.columns:

    transaction_df[
        "Revenue"
    ] = (

        transaction_df[
            "Quantity"
        ]

        *

        transaction_df[
            "UnitPrice"
        ]
    )


    print(
        "\nRevenue column created successfully:"
    )

    print(
        "Revenue = Quantity × UnitPrice"
    )


else:

    print(
        "\nRevenue column already exists."
    )



# ------------------------------------------------------------
# 4.2 Validate Revenue Feature
# ------------------------------------------------------------

print(
    "\nREVENUE VALIDATION"
)


print(
    "Missing Revenue Values:",
    transaction_df[
        "Revenue"
    ].isna().sum()
)


print(
    "Minimum Revenue:",
    transaction_df[
        "Revenue"
    ].min()
)


print(
    "Maximum Revenue:",
    transaction_df[
        "Revenue"
    ].max()
)


print(
    "Total Net Revenue:",
    transaction_df[
        "Revenue"
    ].sum()
)


display(

    transaction_df[
        [
            "InvoiceNo",
            "Quantity",
            "UnitPrice",
            "Revenue"
        ]
    ]
    .head()
)



# ------------------------------------------------------------
# 4.3 Create Positive-Sales Dataset
# ------------------------------------------------------------

# The complete transaction dataset is retained for calculating
# net revenue and cancellation/return metrics.
#
# sales_df represents positive completed sales for customer,
# product, temporal and geographic sales analysis.


sales_df = (
    transaction_df.copy()
)


if "IsCancellation" in sales_df.columns:

    sales_df = (

        sales_df[
            sales_df[
                "IsCancellation"
            ] == 0
        ]
        .copy()
    )


else:

    sales_df = (

        sales_df[
            sales_df[
                "Quantity"
            ] > 0
        ]
        .copy()
    )


# Additional safeguard

sales_df = (

    sales_df[
        sales_df[
            "Revenue"
        ] > 0
    ]
    .copy()
)


print(
    "\nOriginal Transaction Rows:",
    len(
        transaction_df
    )
)


print(
    "Positive Sales Rows:",
    len(
        sales_df
    )
)



# ------------------------------------------------------------
# 4.4 Create Date Features
# ------------------------------------------------------------

sales_df[
    "InvoiceDate"
] = pd.to_datetime(

    sales_df[
        "InvoiceDate"
    ],

    errors="coerce"
)


sales_df[
    "Year"
] = (

    sales_df[
        "InvoiceDate"
    ]
    .dt.year
)


sales_df[
    "Month"
] = (

    sales_df[
        "InvoiceDate"
    ]
    .dt.month
)


sales_df[
    "MonthName"
] = (

    sales_df[
        "InvoiceDate"
    ]
    .dt.month_name()
)


sales_df[
    "YearMonth"
] = (

    sales_df[
        "InvoiceDate"
    ]
    .dt.to_period(
        "M"
    )
    .astype(str)
)


sales_df[
    "Date"
] = (

    sales_df[
        "InvoiceDate"
    ]
    .dt.date
)


sales_df[
    "DayOfWeek"
] = (

    sales_df[
        "InvoiceDate"
    ]
    .dt.day_name()
)


sales_df[
    "Hour"
] = (

    sales_df[
        "InvoiceDate"
    ]
    .dt.hour
)


print(
    "\nDate features created successfully."
)



# ============================================================
# 5. EXECUTIVE KPI ANALYSIS
# ============================================================


# ------------------------------------------------------------
# 5.1 Positive Sales Revenue
# ------------------------------------------------------------

positive_sales_revenue = (

    sales_df[
        "Revenue"
    ]
    .sum()
)



# ------------------------------------------------------------
# 5.2 Net Revenue
# ------------------------------------------------------------

net_revenue = (

    transaction_df[
        "Revenue"
    ]
    .sum()
)



# ------------------------------------------------------------
# 5.3 Total Orders
# ------------------------------------------------------------

total_orders = (

    sales_df[
        "InvoiceNo"
    ]
    .nunique()
)



# ------------------------------------------------------------
# 5.4 Total Customers
# ------------------------------------------------------------

total_customers = (

    sales_df[
        "CustomerID"
    ]
    .nunique()
)



# ------------------------------------------------------------
# 5.5 Unique Products
# ------------------------------------------------------------

total_products = (

    sales_df[
        "StockCode"
    ]
    .nunique()
)



# ------------------------------------------------------------
# 5.6 Units Sold
# ------------------------------------------------------------

total_units = (

    sales_df[
        "Quantity"
    ]
    .sum()
)



# ------------------------------------------------------------
# 5.7 Average Order Value
# ------------------------------------------------------------

average_order_value = (

    positive_sales_revenue
    /
    total_orders

    if total_orders > 0

    else np.nan
)



# ------------------------------------------------------------
# 5.8 Revenue per Customer
# ------------------------------------------------------------

revenue_per_customer = (

    positive_sales_revenue
    /
    total_customers

    if total_customers > 0

    else np.nan
)



# ------------------------------------------------------------
# 5.9 Orders per Customer
# ------------------------------------------------------------

orders_per_customer = (

    total_orders
    /
    total_customers

    if total_customers > 0

    else np.nan
)



# ------------------------------------------------------------
# 5.10 Units per Order
# ------------------------------------------------------------

units_per_order = (

    total_units
    /
    total_orders

    if total_orders > 0

    else np.nan
)



# ------------------------------------------------------------
# 5.11 Cancellation Line Rate
# ------------------------------------------------------------

if "IsCancellation" in transaction_df.columns:

    cancellation_line_count = (

        transaction_df[
            "IsCancellation"
        ]
        .eq(1)
        .sum()
    )


else:

    cancellation_line_count = (

        transaction_df[
            "Quantity"
        ]
        .lt(0)
        .sum()
    )


cancellation_line_rate = (

    cancellation_line_count
    /
    len(
        transaction_df
    )
    * 100

    if len(
        transaction_df
    ) > 0

    else np.nan
)



# ------------------------------------------------------------
# 5.12 Executive KPI Table
# ------------------------------------------------------------

executive_kpis = pd.DataFrame({

    "KPI": [

        "Positive Sales Revenue",

        "Net Revenue",

        "Total Orders",

        "Total Customers",

        "Unique Products",

        "Units Sold",

        "Average Order Value",

        "Revenue per Customer",

        "Orders per Customer",

        "Units per Order",

        "Cancellation Line Rate (%)"
    ],

    "Value": [

        positive_sales_revenue,

        net_revenue,

        total_orders,

        total_customers,

        total_products,

        total_units,

        average_order_value,

        revenue_per_customer,

        orders_per_customer,

        units_per_order,

        cancellation_line_rate
    ]
})


print(
    "\nEXECUTIVE KPI SUMMARY"
)


display(
    executive_kpis.round(2)
)



# ============================================================
# 6. ORDER & SALES PERFORMANCE
# ============================================================


# ------------------------------------------------------------
# 6.1 Create Order-Level Performance Dataset
# ------------------------------------------------------------

order_performance = (

    sales_df

    .groupby(
        "InvoiceNo"
    )

    .agg(

        OrderRevenue=(
            "Revenue",
            "sum"
        ),

        Units=(
            "Quantity",
            "sum"
        ),

        ProductLines=(
            "StockCode",
            "nunique"
        ),

        CustomerID=(
            "CustomerID",
            "first"
        ),

        InvoiceDate=(
            "InvoiceDate",
            "min"
        )
    )

    .reset_index()
)


print(
    "\nORDER PERFORMANCE SUMMARY"
)


display(

    order_performance[
        [
            "OrderRevenue",
            "Units",
            "ProductLines"
        ]
    ]
    .describe()
    .T
)



# ------------------------------------------------------------
# 6.2 Order Revenue Distribution
# ------------------------------------------------------------

fig, ax = plt.subplots(
    figsize=(10, 5)
)


sns.histplot(

    data=order_performance,

    x="OrderRevenue",

    bins=50,

    kde=True,

    ax=ax
)


ax.set_title(
    "Distribution of Order Revenue"
)


ax.set_xlabel(
    "Order Revenue"
)


ax.set_ylabel(
    "Frequency"
)


plt.tight_layout()

plt.show()



# ------------------------------------------------------------
# 6.3 Top 20 Orders by Revenue
# ------------------------------------------------------------

top_orders = (

    order_performance

    .nlargest(
        20,
        "OrderRevenue"
    )
)


print(
    "\nTOP 20 ORDERS BY REVENUE"
)


display(
    top_orders
)



# ============================================================
# 7. TEMPORAL PERFORMANCE ANALYSIS
# ============================================================


# ------------------------------------------------------------
# 7.1 Monthly Performance
# ------------------------------------------------------------

monthly_performance = (

    sales_df

    .groupby(
        "YearMonth"
    )

    .agg(

        Revenue=(
            "Revenue",
            "sum"
        ),

        Orders=(
            "InvoiceNo",
            "nunique"
        ),

        Customers=(
            "CustomerID",
            "nunique"
        ),

        Units=(
            "Quantity",
            "sum"
        ),

        Products=(
            "StockCode",
            "nunique"
        )
    )

    .reset_index()
)



# ------------------------------------------------------------
# 7.2 Monthly Average Order Value
# ------------------------------------------------------------

monthly_performance[
    "AverageOrderValue"
] = (

    monthly_performance[
        "Revenue"
    ]

    /

    monthly_performance[
        "Orders"
    ]
)



# ------------------------------------------------------------
# 7.3 Monthly Revenue per Customer
# ------------------------------------------------------------

monthly_performance[
    "RevenuePerCustomer"
] = (

    monthly_performance[
        "Revenue"
    ]

    /

    monthly_performance[
        "Customers"
    ]
)



# ------------------------------------------------------------
# 7.4 Month-on-Month Revenue Growth
# ------------------------------------------------------------

monthly_performance[
    "RevenueGrowthPct"
] = (

    monthly_performance[
        "Revenue"
    ]
    .pct_change()
    * 100
)



# ------------------------------------------------------------
# 7.5 Month-on-Month Order Growth
# ------------------------------------------------------------

monthly_performance[
    "OrderGrowthPct"
] = (

    monthly_performance[
        "Orders"
    ]
    .pct_change()
    * 100
)


print(
    "\nMONTHLY PERFORMANCE"
)


display(
    monthly_performance.round(2)
)



# ------------------------------------------------------------
# 7.6 Monthly Revenue Trend
# ------------------------------------------------------------

fig, ax = plt.subplots(
    figsize=(12, 6)
)


ax.plot(

    monthly_performance[
        "YearMonth"
    ],

    monthly_performance[
        "Revenue"
    ],

    marker="o"
)


ax.set_title(
    "Monthly Revenue Performance"
)


ax.set_xlabel(
    "Month"
)


ax.set_ylabel(
    "Revenue"
)


plt.xticks(
    rotation=45
)


plt.tight_layout()

plt.show()



# ------------------------------------------------------------
# 7.7 Monthly Orders Trend
# ------------------------------------------------------------

fig, ax = plt.subplots(
    figsize=(12, 6)
)


ax.plot(

    monthly_performance[
        "YearMonth"
    ],

    monthly_performance[
        "Orders"
    ],

    marker="o"
)


ax.set_title(
    "Monthly Order Performance"
)


ax.set_xlabel(
    "Month"
)


ax.set_ylabel(
    "Orders"
)


plt.xticks(
    rotation=45
)


plt.tight_layout()

plt.show()



# ------------------------------------------------------------
# 7.8 Monthly Average Order Value
# ------------------------------------------------------------

fig, ax = plt.subplots(
    figsize=(12, 6)
)


ax.plot(

    monthly_performance[
        "YearMonth"
    ],

    monthly_performance[
        "AverageOrderValue"
    ],

    marker="o"
)


ax.set_title(
    "Monthly Average Order Value"
)


ax.set_xlabel(
    "Month"
)


ax.set_ylabel(
    "Average Order Value"
)


plt.xticks(
    rotation=45
)


plt.tight_layout()

plt.show()



# ------------------------------------------------------------
# 7.9 Daily Performance
# ------------------------------------------------------------

daily_performance = (

    sales_df

    .groupby(
        "Date"
    )

    .agg(

        Revenue=(
            "Revenue",
            "sum"
        ),

        Orders=(
            "InvoiceNo",
            "nunique"
        ),

        Customers=(
            "CustomerID",
            "nunique"
        ),

        Units=(
            "Quantity",
            "sum"
        )
    )

    .reset_index()
)


daily_performance[
    "RevenueRolling7Day"
] = (

    daily_performance[
        "Revenue"
    ]

    .rolling(
        window=7,
        min_periods=1
    )

    .mean()
)



# ------------------------------------------------------------
# 7.10 Daily Revenue Trend
# ------------------------------------------------------------

fig, ax = plt.subplots(
    figsize=(13, 6)
)


ax.plot(

    daily_performance[
        "Date"
    ],

    daily_performance[
        "Revenue"
    ],

    alpha=0.5,

    label="Daily Revenue"
)


ax.plot(

    daily_performance[
        "Date"
    ],

    daily_performance[
        "RevenueRolling7Day"
    ],

    label="7-Day Rolling Average"
)


ax.set_title(
    "Daily Revenue Performance"
)


ax.set_xlabel(
    "Date"
)


ax.set_ylabel(
    "Revenue"
)


ax.legend()


plt.xticks(
    rotation=45
)


plt.tight_layout()

plt.show()



# ------------------------------------------------------------
# 7.11 Day-of-Week Performance
# ------------------------------------------------------------

weekday_order = [

    "Monday",

    "Tuesday",

    "Wednesday",

    "Thursday",

    "Friday",

    "Saturday",

    "Sunday"
]


weekday_performance = (

    sales_df

    .groupby(
        "DayOfWeek"
    )

    .agg(

        Revenue=(
            "Revenue",
            "sum"
        ),

        Orders=(
            "InvoiceNo",
            "nunique"
        ),

        Customers=(
            "CustomerID",
            "nunique"
        ),

        Units=(
            "Quantity",
            "sum"
        )
    )

    .reindex(
        weekday_order
    )

    .dropna(
        how="all"
    )

    .reset_index()
)


weekday_performance[
    "AverageOrderValue"
] = (

    weekday_performance[
        "Revenue"
    ]

    /

    weekday_performance[
        "Orders"
    ]
)


print(
    "\nDAY-OF-WEEK PERFORMANCE"
)


display(
    weekday_performance.round(2)
)



# ------------------------------------------------------------
# 7.12 Revenue by Day of Week
# ------------------------------------------------------------

fig, ax = plt.subplots(
    figsize=(10, 5)
)


sns.barplot(

    data=weekday_performance,

    x="DayOfWeek",

    y="Revenue",

    ax=ax
)


ax.set_title(
    "Revenue by Day of Week"
)


ax.set_xlabel(
    "Day of Week"
)


ax.set_ylabel(
    "Revenue"
)


plt.xticks(
    rotation=30
)


plt.tight_layout()

plt.show()



# ------------------------------------------------------------
# 7.13 Hourly Performance
# ------------------------------------------------------------

hourly_performance = (

    sales_df

    .groupby(
        "Hour"
    )

    .agg(

        Revenue=(
            "Revenue",
            "sum"
        ),

        Orders=(
            "InvoiceNo",
            "nunique"
        ),

        Customers=(
            "CustomerID",
            "nunique"
        ),

        Units=(
            "Quantity",
            "sum"
        )
    )

    .reset_index()
)


hourly_performance[
    "AverageOrderValue"
] = (

    hourly_performance[
        "Revenue"
    ]

    /

    hourly_performance[
        "Orders"
    ]
)


print(
    "\nHOURLY PERFORMANCE"
)


display(
    hourly_performance.round(2)
)



# ------------------------------------------------------------
# 7.14 Hourly Revenue
# ------------------------------------------------------------

fig, ax = plt.subplots(
    figsize=(11, 5)
)


sns.lineplot(

    data=hourly_performance,

    x="Hour",

    y="Revenue",

    marker="o",

    ax=ax
)


ax.set_title(
    "Revenue by Hour of Day"
)


ax.set_xlabel(
    "Hour"
)


ax.set_ylabel(
    "Revenue"
)


plt.tight_layout()

plt.show()



# ------------------------------------------------------------
# 7.15 Day × Hour Revenue Heatmap
# ------------------------------------------------------------

day_hour_revenue = (

    sales_df

    .pivot_table(

        values="Revenue",

        index="DayOfWeek",

        columns="Hour",

        aggfunc="sum",

        fill_value=0
    )

    .reindex(
        weekday_order
    )

    .dropna(
        how="all"
    )
)


fig, ax = plt.subplots(
    figsize=(14, 6)
)


sns.heatmap(

    day_hour_revenue,

    ax=ax
)


ax.set_title(
    "Revenue Heatmap: Day of Week × Hour"
)


ax.set_xlabel(
    "Hour"
)


ax.set_ylabel(
    "Day of Week"
)


plt.tight_layout()

plt.show()



# ============================================================
# 8. CUSTOMER PERFORMANCE ANALYSIS
# ============================================================


# ------------------------------------------------------------
# 8.1 Create Customer Performance Dataset
# ------------------------------------------------------------

customer_performance = (

    sales_df

    .groupby(
        "CustomerID"
    )

    .agg(

        TotalRevenue=(
            "Revenue",
            "sum"
        ),

        OrderCount=(
            "InvoiceNo",
            "nunique"
        ),

        TotalQuantity=(
            "Quantity",
            "sum"
        ),

        UniqueProducts=(
            "StockCode",
            "nunique"
        ),

        FirstPurchase=(
            "InvoiceDate",
            "min"
        ),

        LastPurchase=(
            "InvoiceDate",
            "max"
        )
    )

    .reset_index()
)



# ------------------------------------------------------------
# 8.2 Average Order Value by Customer
# ------------------------------------------------------------

customer_performance[
    "AverageOrderValue"
] = (

    customer_performance[
        "TotalRevenue"
    ]

    /

    customer_performance[
        "OrderCount"
    ]
)



# ------------------------------------------------------------
# 8.3 Customer Active Period
# ------------------------------------------------------------

customer_performance[
    "ActiveDays"
] = (

    customer_performance[
        "LastPurchase"
    ]

    -

    customer_performance[
        "FirstPurchase"
    ]
).dt.days



# ------------------------------------------------------------
# 8.4 Customer Summary Statistics
# ------------------------------------------------------------

print(
    "\nCUSTOMER PERFORMANCE SUMMARY"
)


display(

    customer_performance[
        [
            "TotalRevenue",
            "OrderCount",
            "TotalQuantity",
            "UniqueProducts",
            "AverageOrderValue",
            "ActiveDays"
        ]
    ]

    .describe()

    .T
)



# ------------------------------------------------------------
# 8.5 Top 20 Customers by Revenue
# ------------------------------------------------------------

top_customers = (

    customer_performance

    .nlargest(
        20,
        "TotalRevenue"
    )
)


print(
    "\nTOP 20 CUSTOMERS BY REVENUE"
)


display(
    top_customers
)



# ------------------------------------------------------------
# 8.6 Top Customer Revenue Visualisation
# ------------------------------------------------------------

top_customer_plot = (

    top_customers

    .sort_values(
        "TotalRevenue"
    )
)


fig, ax = plt.subplots(
    figsize=(10, 8)
)


ax.barh(

    top_customer_plot[
        "CustomerID"
    ].astype(str),

    top_customer_plot[
        "TotalRevenue"
    ]
)


ax.set_title(
    "Top 20 Customers by Revenue"
)


ax.set_xlabel(
    "Revenue"
)


ax.set_ylabel(
    "Customer ID"
)


plt.tight_layout()

plt.show()



# ============================================================
# 9. CUSTOMER SEGMENT PERFORMANCE
# ============================================================


# ------------------------------------------------------------
# 9.1 Validate CustomerID
# ------------------------------------------------------------

if "CustomerID" not in segment_df.columns:

    raise ValueError(
        "CustomerID is missing from customer_segments.csv."
    )



# ------------------------------------------------------------
# 9.2 Determine Segment Column
# ------------------------------------------------------------

if "Segment" in segment_df.columns:

    segment_column = "Segment"


elif "Cluster" in segment_df.columns:

    segment_column = "Cluster"


else:

    raise ValueError(

        "Neither Segment nor Cluster was found "
        "in customer_segments.csv."
    )


print(
    "\nSegment Column Used:",
    segment_column
)



# ------------------------------------------------------------
# 9.3 Merge Customer Performance with Segments
# ------------------------------------------------------------

customer_segment_base = (

    customer_performance

    .merge(

        segment_df[
            [
                "CustomerID",
                segment_column
            ]
        ]
        .drop_duplicates(
            subset="CustomerID"
        ),

        on="CustomerID",

        how="left"
    )
)


print(
    "Customers without segment assignment:",
    customer_segment_base[
        segment_column
    ]
    .isna()
    .sum()
)



# ------------------------------------------------------------
# 9.4 Segment Performance Table
# ------------------------------------------------------------

segment_performance = (

    customer_segment_base

    .groupby(
        segment_column,
        dropna=False
    )

    .agg(

        Customers=(
            "CustomerID",
            "nunique"
        ),

        TotalRevenue=(
            "TotalRevenue",
            "sum"
        ),

        MeanRevenuePerCustomer=(
            "TotalRevenue",
            "mean"
        ),

        MedianRevenuePerCustomer=(
            "TotalRevenue",
            "median"
        ),

        MeanOrders=(
            "OrderCount",
            "mean"
        ),

        MeanAverageOrderValue=(
            "AverageOrderValue",
            "mean"
        ),

        MeanTotalQuantity=(
            "TotalQuantity",
            "mean"
        ),

        MeanUniqueProducts=(
            "UniqueProducts",
            "mean"
        )
    )

    .reset_index()
)



# ------------------------------------------------------------
# 9.5 Customer Percentage
# ------------------------------------------------------------

segment_performance[
    "CustomerPercentage"
] = (

    segment_performance[
        "Customers"
    ]

    /

    segment_performance[
        "Customers"
    ].sum()

    * 100
)



# ------------------------------------------------------------
# 9.6 Revenue Percentage
# ------------------------------------------------------------

segment_performance[
    "RevenuePercentage"
] = (

    segment_performance[
        "TotalRevenue"
    ]

    /

    segment_performance[
        "TotalRevenue"
    ].sum()

    * 100
)


print(
    "\nCUSTOMER SEGMENT PERFORMANCE"
)


display(
    segment_performance.round(2)
)



# ------------------------------------------------------------
# 9.7 Segment Revenue Contribution
# ------------------------------------------------------------

fig, ax = plt.subplots(
    figsize=(10, 6)
)


sns.barplot(

    data=segment_performance,

    x=segment_column,

    y="TotalRevenue",

    ax=ax
)


ax.set_title(
    "Revenue Contribution by Customer Segment"
)


ax.set_xlabel(
    "Customer Segment"
)


ax.set_ylabel(
    "Total Revenue"
)


plt.xticks(
    rotation=30,
    ha="right"
)


plt.tight_layout()

plt.show()



# ------------------------------------------------------------
# 9.8 Customer Distribution by Segment
# ------------------------------------------------------------

fig, ax = plt.subplots(
    figsize=(10, 6)
)


sns.barplot(

    data=segment_performance,

    x=segment_column,

    y="Customers",

    ax=ax
)


ax.set_title(
    "Customer Distribution by Segment"
)


ax.set_xlabel(
    "Customer Segment"
)


ax.set_ylabel(
    "Customers"
)


plt.xticks(
    rotation=30,
    ha="right"
)


plt.tight_layout()

plt.show()



# ============================================================
# 10. PRODUCT PERFORMANCE ANALYSIS
# ============================================================


# ------------------------------------------------------------
# 10.1 Product-Level Business Performance
# ------------------------------------------------------------

product_performance = (

    sales_df

    .groupby(
        "StockCode"
    )

    .agg(

        TotalRevenue=(
            "Revenue",
            "sum"
        ),

        TotalQuantity=(
            "Quantity",
            "sum"
        ),

        Orders=(
            "InvoiceNo",
            "nunique"
        ),

        Customers=(
            "CustomerID",
            "nunique"
        ),

        AverageUnitPrice=(
            "UnitPrice",
            "mean"
        )
    )

    .reset_index()
)



# ------------------------------------------------------------
# 10.2 Add Product Description
# ------------------------------------------------------------

if "Description" in sales_df.columns:

    product_description = (

        sales_df

        .groupby(
            "StockCode"
        )[
            "Description"
        ]

        .agg(

            lambda x:

            x.mode().iloc[0]

            if not x.mode().empty

            else x.iloc[0]
        )

        .reset_index()
    )


    product_performance = (

        product_performance

        .merge(

            product_description,

            on="StockCode",

            how="left"
        )
    )



# ------------------------------------------------------------
# 10.3 Revenue Contribution Percentage
# ------------------------------------------------------------

product_performance[
    "RevenuePercentage"
] = (

    product_performance[
        "TotalRevenue"
    ]

    /

    product_performance[
        "TotalRevenue"
    ].sum()

    * 100
)



# ------------------------------------------------------------
# 10.4 Top Products by Revenue
# ------------------------------------------------------------

top_products_revenue = (

    product_performance

    .nlargest(
        20,
        "TotalRevenue"
    )
)


print(
    "\nTOP 20 PRODUCTS BY REVENUE"
)


display(
    top_products_revenue
)



# ------------------------------------------------------------
# 10.5 Top Products by Quantity
# ------------------------------------------------------------

top_products_quantity = (

    product_performance

    .nlargest(
        20,
        "TotalQuantity"
    )
)


print(
    "\nTOP 20 PRODUCTS BY QUANTITY"
)


display(
    top_products_quantity
)



# ------------------------------------------------------------
# 10.6 Top Products by Customer Reach
# ------------------------------------------------------------

top_products_customers = (

    product_performance

    .nlargest(
        20,
        "Customers"
    )
)


print(
    "\nTOP 20 PRODUCTS BY CUSTOMER REACH"
)


display(
    top_products_customers
)



# ------------------------------------------------------------
# 10.7 Top Revenue Products Visualisation
# ------------------------------------------------------------

top_product_plot = (

    top_products_revenue

    .sort_values(
        "TotalRevenue"
    )
)


if "Description" in top_product_plot.columns:

    product_label_column = (
        "Description"
    )


else:

    product_label_column = (
        "StockCode"
    )


fig, ax = plt.subplots(
    figsize=(11, 8)
)


ax.barh(

    top_product_plot[
        product_label_column
    ].astype(str),

    top_product_plot[
        "TotalRevenue"
    ]
)


ax.set_title(
    "Top 20 Products by Revenue"
)


ax.set_xlabel(
    "Revenue"
)


ax.set_ylabel(
    "Product"
)


plt.tight_layout()

plt.show()



# ============================================================
# 11. GEOGRAPHIC PERFORMANCE ANALYSIS
# ============================================================


# ------------------------------------------------------------
# 11.1 Country Performance
# ------------------------------------------------------------

country_performance = (

    sales_df

    .groupby(
        "Country"
    )

    .agg(

        Revenue=(
            "Revenue",
            "sum"
        ),

        Orders=(
            "InvoiceNo",
            "nunique"
        ),

        Customers=(
            "CustomerID",
            "nunique"
        ),

        Units=(
            "Quantity",
            "sum"
        ),

        Products=(
            "StockCode",
            "nunique"
        )
    )

    .reset_index()
)



# ------------------------------------------------------------
# 11.2 Country AOV
# ------------------------------------------------------------

country_performance[
    "AverageOrderValue"
] = (

    country_performance[
        "Revenue"
    ]

    /

    country_performance[
        "Orders"
    ]
)



# ------------------------------------------------------------
# 11.3 Revenue per Customer
# ------------------------------------------------------------

country_performance[
    "RevenuePerCustomer"
] = (

    country_performance[
        "Revenue"
    ]

    /

    country_performance[
        "Customers"
    ]
)



# ------------------------------------------------------------
# 11.4 Revenue Contribution
# ------------------------------------------------------------

country_performance[
    "RevenuePercentage"
] = (

    country_performance[
        "Revenue"
    ]

    /

    country_performance[
        "Revenue"
    ].sum()

    * 100
)


country_performance = (

    country_performance

    .sort_values(
        "Revenue",
        ascending=False
    )

    .reset_index(
        drop=True
    )
)


print(
    "\nCOUNTRY PERFORMANCE"
)


display(
    country_performance.round(2)
)



# ------------------------------------------------------------
# 11.5 Top Countries by Revenue
# ------------------------------------------------------------

top_countries = (

    country_performance

    .head(15)

    .sort_values(
        "Revenue"
    )
)


fig, ax = plt.subplots(
    figsize=(10, 7)
)


ax.barh(

    top_countries[
        "Country"
    ],

    top_countries[
        "Revenue"
    ]
)


ax.set_title(
    "Top 15 Countries by Revenue"
)


ax.set_xlabel(
    "Revenue"
)


ax.set_ylabel(
    "Country"
)


plt.tight_layout()

plt.show()



# ============================================================
# 12. CANCELLATION / RETURN ANALYSIS
# ============================================================


# ------------------------------------------------------------
# 12.1 Identify Cancellation Records
# ------------------------------------------------------------

if "IsCancellation" in transaction_df.columns:

    cancellation_df = (

        transaction_df[
            transaction_df[
                "IsCancellation"
            ] == 1
        ]

        .copy()
    )


else:

    cancellation_df = (

        transaction_df[
            transaction_df[
                "Quantity"
            ] < 0
        ]

        .copy()
    )


print(
    "\nCancellation Records:",
    len(
        cancellation_df
    )
)



# ------------------------------------------------------------
# 12.2 Cancellation KPIs
# ------------------------------------------------------------

cancelled_invoices = (

    cancellation_df[
        "InvoiceNo"
    ]
    .nunique()
)


cancelled_customers = (

    cancellation_df[
        "CustomerID"
    ]
    .nunique()
)


cancelled_quantity = (

    cancellation_df[
        "Quantity"
    ]
    .abs()
    .sum()
)


cancelled_value = (

    cancellation_df[
        "Revenue"
    ]
    .abs()
    .sum()
)


total_unique_invoices = (

    transaction_df[
        "InvoiceNo"
    ]
    .nunique()
)


invoice_cancellation_rate = (

    cancelled_invoices

    /

    total_unique_invoices

    * 100

    if total_unique_invoices > 0

    else np.nan
)



# ------------------------------------------------------------
# 12.3 Cancellation KPI Table
# ------------------------------------------------------------

cancellation_kpis = pd.DataFrame({

    "KPI": [

        "Cancellation Records",

        "Cancelled Invoices",

        "Customers with Cancellations",

        "Cancelled Units",

        "Absolute Cancelled Value",

        "Invoice Cancellation Rate (%)"
    ],

    "Value": [

        len(
            cancellation_df
        ),

        cancelled_invoices,

        cancelled_customers,

        cancelled_quantity,

        cancelled_value,

        invoice_cancellation_rate
    ]
})


print(
    "\nCANCELLATION KPI SUMMARY"
)


display(
    cancellation_kpis.round(2)
)



# ------------------------------------------------------------
# 12.4 Cancellation by Country
# ------------------------------------------------------------

cancellation_country = pd.DataFrame()


if len(
    cancellation_df
) > 0:

    cancellation_country = (

        cancellation_df

        .groupby(
            "Country"
        )

        .agg(

            CancellationRecords=(
                "InvoiceNo",
                "size"
            ),

            CancelledInvoices=(
                "InvoiceNo",
                "nunique"
            ),

            CancelledValue=(

                "Revenue",

                lambda x:
                x.abs().sum()
            )
        )

        .reset_index()

        .sort_values(
            "CancelledValue",
            ascending=False
        )
    )


    print(
        "\nCANCELLATION BY COUNTRY"
    )


    display(
        cancellation_country.head(15)
    )



# ------------------------------------------------------------
# 12.5 Cancellation by Product
# ------------------------------------------------------------

cancellation_product = pd.DataFrame()


if len(
    cancellation_df
) > 0:

    cancellation_product = (

        cancellation_df

        .groupby(
            "StockCode"
        )

        .agg(

            CancellationRecords=(
                "InvoiceNo",
                "size"
            ),

            CancelledQuantity=(

                "Quantity",

                lambda x:
                x.abs().sum()
            ),

            CancelledValue=(

                "Revenue",

                lambda x:
                x.abs().sum()
            )
        )

        .reset_index()

        .sort_values(
            "CancelledValue",
            ascending=False
        )
    )


    print(
        "\nCANCELLATION BY PRODUCT"
    )


    display(
        cancellation_product.head(20)
    )



# ------------------------------------------------------------
# 12.6 Monthly Cancellation Trend
# ------------------------------------------------------------

monthly_cancellations = pd.DataFrame()


if len(
    cancellation_df
) > 0:

    cancellation_df[
        "InvoiceDate"
    ] = pd.to_datetime(

        cancellation_df[
            "InvoiceDate"
        ],

        errors="coerce"
    )


    cancellation_df[
        "YearMonth"
    ] = (

        cancellation_df[
            "InvoiceDate"
        ]

        .dt.to_period(
            "M"
        )

        .astype(str)
    )


    monthly_cancellations = (

        cancellation_df

        .groupby(
            "YearMonth"
        )

        .agg(

            CancellationRecords=(
                "InvoiceNo",
                "size"
            ),

            CancelledInvoices=(
                "InvoiceNo",
                "nunique"
            ),

            CancelledValue=(

                "Revenue",

                lambda x:
                x.abs().sum()
            )
        )

        .reset_index()
    )


    print(
        "\nMONTHLY CANCELLATION PERFORMANCE"
    )


    display(
        monthly_cancellations.round(2)
    )



# ============================================================
# 13. CUSTOMER REVENUE CONCENTRATION / PARETO ANALYSIS
# ============================================================


# ------------------------------------------------------------
# 13.1 Prepare Customer Pareto Dataset
# ------------------------------------------------------------

customer_pareto = (

    customer_performance[
        [
            "CustomerID",
            "TotalRevenue"
        ]
    ]

    .sort_values(
        "TotalRevenue",
        ascending=False
    )

    .reset_index(
        drop=True
    )
)



# ------------------------------------------------------------
# 13.2 Cumulative Revenue
# ------------------------------------------------------------

customer_pareto[
    "CumulativeRevenue"
] = (

    customer_pareto[
        "TotalRevenue"
    ]
    .cumsum()
)



# ------------------------------------------------------------
# 13.3 Cumulative Revenue Percentage
# ------------------------------------------------------------

customer_pareto[
    "CumulativeRevenuePct"
] = (

    customer_pareto[
        "CumulativeRevenue"
    ]

    /

    customer_pareto[
        "TotalRevenue"
    ].sum()

    * 100
)



# ------------------------------------------------------------
# 13.4 Cumulative Customer Percentage
# ------------------------------------------------------------

customer_pareto[
    "CumulativeCustomerPct"
] = (

    np.arange(
        1,
        len(
            customer_pareto
        ) + 1
    )

    /

    len(
        customer_pareto
    )

    * 100
)


display(
    customer_pareto.head(20)
)



# ------------------------------------------------------------
# 13.5 Top 20% Customer Revenue Contribution
# ------------------------------------------------------------

top_20_customer_count = max(

    1,

    int(

        np.ceil(

            len(
                customer_pareto
            )

            * 0.20
        )
    )
)


top_20_customer_revenue = (

    customer_pareto

    .head(
        top_20_customer_count
    )[
        "TotalRevenue"
    ]

    .sum()
)


top_20_customer_revenue_pct = (

    top_20_customer_revenue

    /

    customer_pareto[
        "TotalRevenue"
    ].sum()

    * 100
)


print(
    "\nRevenue generated by top 20% of customers:"
)


print(
    f"{top_20_customer_revenue_pct:.2f}%"
)



# ------------------------------------------------------------
# 13.6 Customer Pareto Curve
# ------------------------------------------------------------

fig, ax = plt.subplots(
    figsize=(10, 6)
)


ax.plot(

    customer_pareto[
        "CumulativeCustomerPct"
    ],

    customer_pareto[
        "CumulativeRevenuePct"
    ]
)


ax.axvline(
    x=20,
    linestyle="--"
)


ax.axhline(
    y=80,
    linestyle="--"
)


ax.set_title(
    "Customer Revenue Concentration"
)


ax.set_xlabel(
    "Cumulative Percentage of Customers"
)


ax.set_ylabel(
    "Cumulative Percentage of Revenue"
)


plt.tight_layout()

plt.show()



# ============================================================
# 14. PRODUCT REVENUE CONCENTRATION / PARETO ANALYSIS
# ============================================================


# ------------------------------------------------------------
# 14.1 Prepare Product Pareto Dataset
# ------------------------------------------------------------

product_pareto = (

    product_performance[
        [
            "StockCode",
            "TotalRevenue"
        ]
    ]

    .sort_values(
        "TotalRevenue",
        ascending=False
    )

    .reset_index(
        drop=True
    )
)



# ------------------------------------------------------------
# 14.2 Cumulative Revenue
# ------------------------------------------------------------

product_pareto[
    "CumulativeRevenue"
] = (

    product_pareto[
        "TotalRevenue"
    ]
    .cumsum()
)



# ------------------------------------------------------------
# 14.3 Cumulative Revenue Percentage
# ------------------------------------------------------------

product_pareto[
    "CumulativeRevenuePct"
] = (

    product_pareto[
        "CumulativeRevenue"
    ]

    /

    product_pareto[
        "TotalRevenue"
    ].sum()

    * 100
)



# ------------------------------------------------------------
# 14.4 Cumulative Product Percentage
# ------------------------------------------------------------

product_pareto[
    "CumulativeProductPct"
] = (

    np.arange(
        1,
        len(
            product_pareto
        ) + 1
    )

    /

    len(
        product_pareto
    )

    * 100
)



# ------------------------------------------------------------
# 14.5 Top 20% Product Revenue Contribution
# ------------------------------------------------------------

top_20_product_count = max(

    1,

    int(

        np.ceil(

            len(
                product_pareto
            )

            * 0.20
        )
    )
)


top_20_product_revenue = (

    product_pareto

    .head(
        top_20_product_count
    )[
        "TotalRevenue"
    ]

    .sum()
)


top_20_product_revenue_pct = (

    top_20_product_revenue

    /

    product_pareto[
        "TotalRevenue"
    ].sum()

    * 100
)


print(
    "\nRevenue generated by top 20% of products:"
)


print(
    f"{top_20_product_revenue_pct:.2f}%"
)



# ------------------------------------------------------------
# 14.6 Product Pareto Curve
# ------------------------------------------------------------

fig, ax = plt.subplots(
    figsize=(10, 6)
)


ax.plot(

    product_pareto[
        "CumulativeProductPct"
    ],

    product_pareto[
        "CumulativeRevenuePct"
    ]
)


ax.axvline(
    x=20,
    linestyle="--"
)


ax.axhline(
    y=80,
    linestyle="--"
)


ax.set_title(
    "Product Revenue Concentration"
)


ax.set_xlabel(
    "Cumulative Percentage of Products"
)


ax.set_ylabel(
    "Cumulative Percentage of Revenue"
)


plt.tight_layout()

plt.show()



# ============================================================
# 15. CROSS-DIMENSIONAL BUSINESS ANALYSIS
# ============================================================


# ------------------------------------------------------------
# 15.1 Attach Segment Labels to Sales Transactions
# ------------------------------------------------------------

segment_lookup = (

    segment_df[
        [
            "CustomerID",
            segment_column
        ]
    ]

    .drop_duplicates(
        subset="CustomerID"
    )
)


transaction_segment_df = (

    sales_df

    .merge(

        segment_lookup,

        on="CustomerID",

        how="left"
    )
)



# ------------------------------------------------------------
# 15.2 Segment × Country Performance
# ------------------------------------------------------------

segment_country_performance = (

    transaction_segment_df

    .groupby(
        [
            segment_column,
            "Country"
        ],
        dropna=False
    )

    .agg(

        Revenue=(
            "Revenue",
            "sum"
        ),

        Orders=(
            "InvoiceNo",
            "nunique"
        ),

        Customers=(
            "CustomerID",
            "nunique"
        ),

        Units=(
            "Quantity",
            "sum"
        )
    )

    .reset_index()
)


print(
    "\nSEGMENT × COUNTRY PERFORMANCE"
)


display(

    segment_country_performance

    .sort_values(
        "Revenue",
        ascending=False
    )

    .head(30)
)



# ------------------------------------------------------------
# 15.3 Segment × Product Performance
# ------------------------------------------------------------

segment_product_performance = (

    transaction_segment_df

    .groupby(
        [
            segment_column,
            "StockCode"
        ],
        dropna=False
    )

    .agg(

        Revenue=(
            "Revenue",
            "sum"
        ),

        Quantity=(
            "Quantity",
            "sum"
        ),

        Orders=(
            "InvoiceNo",
            "nunique"
        ),

        Customers=(
            "CustomerID",
            "nunique"
        )
    )

    .reset_index()
)


print(
    "\nSEGMENT × PRODUCT PERFORMANCE"
)


display(

    segment_product_performance

    .sort_values(
        "Revenue",
        ascending=False
    )

    .head(30)
)



# ------------------------------------------------------------
# 15.4 Segment × Month Performance
# ------------------------------------------------------------

segment_month_performance = (

    transaction_segment_df

    .groupby(
        [
            "YearMonth",
            segment_column
        ],
        dropna=False
    )

    .agg(

        Revenue=(
            "Revenue",
            "sum"
        ),

        Orders=(
            "InvoiceNo",
            "nunique"
        ),

        Customers=(
            "CustomerID",
            "nunique"
        ),

        Units=(
            "Quantity",
            "sum"
        )
    )

    .reset_index()
)


print(
    "\nSEGMENT × MONTH PERFORMANCE"
)


display(
    segment_month_performance.head(30)
)



# ============================================================
# 16. KEY BUSINESS FINDINGS
# ============================================================


# ------------------------------------------------------------
# 16.1 Peak Revenue Month
# ------------------------------------------------------------

best_month_row = (

    monthly_performance

    .loc[
        monthly_performance[
            "Revenue"
        ].idxmax()
    ]
)


best_month = (

    best_month_row[
        "YearMonth"
    ]
)


best_month_revenue = (

    best_month_row[
        "Revenue"
    ]
)



# ------------------------------------------------------------
# 16.2 Highest-Revenue Country
# ------------------------------------------------------------

top_country_row = (

    country_performance

    .iloc[0]
)


top_country = (

    top_country_row[
        "Country"
    ]
)


top_country_revenue = (

    top_country_row[
        "Revenue"
    ]
)


top_country_revenue_share = (

    top_country_row[
        "RevenuePercentage"
    ]
)



# ------------------------------------------------------------
# 16.3 Highest-Revenue Product
# ------------------------------------------------------------

top_product_row = (

    product_performance

    .loc[
        product_performance[
            "TotalRevenue"
        ].idxmax()
    ]
)


top_product_code = (

    top_product_row[
        "StockCode"
    ]
)


top_product_revenue = (

    top_product_row[
        "TotalRevenue"
    ]
)



# ------------------------------------------------------------
# 16.4 Highest-Revenue Customer Segment
# ------------------------------------------------------------

top_segment_row = (

    segment_performance

    .loc[
        segment_performance[
            "TotalRevenue"
        ].idxmax()
    ]
)


top_segment = (

    top_segment_row[
        segment_column
    ]
)


top_segment_revenue_share = (

    top_segment_row[
        "RevenuePercentage"
    ]
)



# ------------------------------------------------------------
# 16.5 Create Structured Business Findings
# ------------------------------------------------------------

business_findings = pd.DataFrame({

    "Finding": [

        "Peak Revenue Month",

        "Highest-Revenue Country",

        "Highest-Revenue Product",

        "Highest-Revenue Customer Segment",

        "Customer Revenue Concentration",

        "Product Revenue Concentration",

        "Invoice Cancellation Rate"
    ],

    "Evidence": [

        (
            f"{best_month} generated "
            f"{best_month_revenue:,.2f} "
            f"in positive sales revenue."
        ),

        (
            f"{top_country} generated "
            f"{top_country_revenue:,.2f}, "
            f"representing "
            f"{top_country_revenue_share:.2f}% "
            f"of positive sales revenue."
        ),

        (
            f"Product {top_product_code} generated "
            f"{top_product_revenue:,.2f} "
            f"in positive sales revenue."
        ),

        (
            f"{top_segment} contributed "
            f"{top_segment_revenue_share:.2f}% "
            f"of analysed customer revenue."
        ),

        (
            f"The top 20% of customers generated "
            f"{top_20_customer_revenue_pct:.2f}% "
            f"of positive customer revenue."
        ),

        (
            f"The top 20% of products generated "
            f"{top_20_product_revenue_pct:.2f}% "
            f"of positive product revenue."
        ),

        (
            f"The invoice cancellation rate was "
            f"{invoice_cancellation_rate:.2f}%."
        )
    ]
})


print(
    "\nKEY BUSINESS FINDINGS"
)


display(
    business_findings
)



# ============================================================
# 17. DASHBOARD KPI TABLES
# ============================================================


# ------------------------------------------------------------
# 17.1 Main Dashboard Summary
# ------------------------------------------------------------

dashboard_summary = pd.DataFrame({

    "Metric": [

        "Positive Sales Revenue",

        "Net Revenue",

        "Orders",

        "Customers",

        "Products",

        "Units",

        "Average Order Value",

        "Revenue per Customer",

        "Invoice Cancellation Rate"
    ],

    "Value": [

        positive_sales_revenue,

        net_revenue,

        total_orders,

        total_customers,

        total_products,

        total_units,

        average_order_value,

        revenue_per_customer,

        invoice_cancellation_rate
    ]
})


print(
    "\nDASHBOARD SUMMARY"
)


display(
    dashboard_summary.round(2)
)



# ============================================================
# 18. EXPORT STREAMLIT-READY DATASETS
# ============================================================


# ------------------------------------------------------------
# 18.1 Create Dashboard Directory
# ------------------------------------------------------------

dashboard_path = Path(
    "../Data/dashboard"
)


dashboard_path.mkdir(
    parents=True,
    exist_ok=True
)



# ------------------------------------------------------------
# 18.2 Executive KPIs
# ------------------------------------------------------------

executive_kpis.to_csv(

    dashboard_path
    /
    "executive_kpis.csv",

    index=False
)



# ------------------------------------------------------------
# 18.3 Dashboard Summary
# ------------------------------------------------------------

dashboard_summary.to_csv(

    dashboard_path
    /
    "dashboard_summary.csv",

    index=False
)



# ------------------------------------------------------------
# 18.4 Monthly Performance
# ------------------------------------------------------------

monthly_performance.to_csv(

    dashboard_path
    /
    "monthly_performance.csv",

    index=False
)



# ------------------------------------------------------------
# 18.5 Daily Performance
# ------------------------------------------------------------

daily_performance.to_csv(

    dashboard_path
    /
    "daily_performance.csv",

    index=False
)



# ------------------------------------------------------------
# 18.6 Weekday Performance
# ------------------------------------------------------------

weekday_performance.to_csv(

    dashboard_path
    /
    "weekday_performance.csv",

    index=False
)



# ------------------------------------------------------------
# 18.7 Hourly Performance
# ------------------------------------------------------------

hourly_performance.to_csv(

    dashboard_path
    /
    "hourly_performance.csv",

    index=False
)



# ------------------------------------------------------------
# 18.8 Order Performance
# ------------------------------------------------------------

order_performance.to_csv(

    dashboard_path
    /
    "order_performance.csv",

    index=False
)



# ------------------------------------------------------------
# 18.9 Customer Performance
# ------------------------------------------------------------

customer_performance.to_csv(

    dashboard_path
    /
    "customer_performance.csv",

    index=False
)



# ------------------------------------------------------------
# 18.10 Customer Segment Performance
# ------------------------------------------------------------

segment_performance.to_csv(

    dashboard_path
    /
    "segment_performance.csv",

    index=False
)



# ------------------------------------------------------------
# 18.11 Product Performance
# ------------------------------------------------------------

product_performance.to_csv(

    dashboard_path
    /
    "product_performance.csv",

    index=False
)



# ------------------------------------------------------------
# 18.12 Country Performance
# ------------------------------------------------------------

country_performance.to_csv(

    dashboard_path
    /
    "country_performance.csv",

    index=False
)



# ------------------------------------------------------------
# 18.13 Customer Pareto
# ------------------------------------------------------------

customer_pareto.to_csv(

    dashboard_path
    /
    "customer_pareto.csv",

    index=False
)



# ------------------------------------------------------------
# 18.14 Product Pareto
# ------------------------------------------------------------

product_pareto.to_csv(

    dashboard_path
    /
    "product_pareto.csv",

    index=False
)



# ------------------------------------------------------------
# 18.15 Segment × Country
# ------------------------------------------------------------

segment_country_performance.to_csv(

    dashboard_path
    /
    "segment_country_performance.csv",

    index=False
)



# ------------------------------------------------------------
# 18.16 Segment × Product
# ------------------------------------------------------------

segment_product_performance.to_csv(

    dashboard_path
    /
    "segment_product_performance.csv",

    index=False
)



# ------------------------------------------------------------
# 18.17 Segment × Month
# ------------------------------------------------------------

segment_month_performance.to_csv(

    dashboard_path
    /
    "segment_month_performance.csv",

    index=False
)



# ------------------------------------------------------------
# 18.18 Cancellation KPIs
# ------------------------------------------------------------

cancellation_kpis.to_csv(

    dashboard_path
    /
    "cancellation_kpis.csv",

    index=False
)



# ------------------------------------------------------------
# 18.19 Cancellation by Country
# ------------------------------------------------------------

if not cancellation_country.empty:

    cancellation_country.to_csv(

        dashboard_path
        /
        "cancellation_by_country.csv",

        index=False
    )



# ------------------------------------------------------------
# 18.20 Cancellation by Product
# ------------------------------------------------------------

if not cancellation_product.empty:

    cancellation_product.to_csv(

        dashboard_path
        /
        "cancellation_by_product.csv",

        index=False
    )



# ------------------------------------------------------------
# 18.21 Monthly Cancellations
# ------------------------------------------------------------

if not monthly_cancellations.empty:

    monthly_cancellations.to_csv(

        dashboard_path
        /
        "monthly_cancellations.csv",

        index=False
    )



# ------------------------------------------------------------
# 18.22 Business Findings
# ------------------------------------------------------------

business_findings.to_csv(

    dashboard_path
    /
    "business_findings.csv",

    index=False
)



# ------------------------------------------------------------
# 18.23 Segmented Customers
# ------------------------------------------------------------

segment_df.to_csv(

    dashboard_path
    /
    "customer_segments.csv",

    index=False
)



# ------------------------------------------------------------
# 18.24 Transaction-Segment Dataset
# ------------------------------------------------------------

transaction_segment_df.to_csv(

    dashboard_path
    /
    "transaction_segments.csv",

    index=False
)


print(
    "\nAll Streamlit-ready datasets exported successfully."
)



# ============================================================
# 19. EXPORT VALIDATION
# ============================================================


exported_files = sorted(

    [
        file.name

        for file
        in dashboard_path.glob(
            "*.csv"
        )
    ]
)


print(
    "\nDASHBOARD FILES CREATED"
)


for number, file in enumerate(
    exported_files,
    start=1
):

    print(
        f"{number}. {file}"
    )


print(
    "\nTotal Dashboard Files:",
    len(
        exported_files
    )
)



# ============================================================
# 20. FINAL NOTEBOOK SUMMARY
# ============================================================


notebook_summary = pd.DataFrame({

    "AnalysisArea": [

        "Data Validation",

        "Revenue Engineering",

        "Executive KPIs",

        "Order Performance",

        "Temporal Analysis",

        "Customer Analysis",

        "Customer Segmentation",

        "Product Analysis",

        "Geographic Analysis",

        "Cancellation Analysis",

        "Customer Pareto Analysis",

        "Product Pareto Analysis",

        "Cross-Dimensional Analysis",

        "Business Findings",

        "Dashboard Data Export"
    ],

    "Status": [

        "Completed",

        "Completed",

        "Completed",

        "Completed",

        "Completed",

        "Completed",

        "Completed",

        "Completed",

        "Completed",

        "Completed",

        "Completed",

        "Completed",

        "Completed",

        "Completed",

        "Completed"
    ]
})


display(
    notebook_summary
)



# ------------------------------------------------------------
# 20.1 Final Analytical Summary
# ------------------------------------------------------------

print(
    "\n"
    + "=" * 60
)


print(
    "NOTEBOOK 06 FINAL SUMMARY"
)


print(
    "=" * 60
)


print(
    f"Positive Sales Revenue : "
    f"{positive_sales_revenue:,.2f}"
)


print(
    f"Net Revenue            : "
    f"{net_revenue:,.2f}"
)


print(
    f"Orders                 : "
    f"{total_orders:,}"
)


print(
    f"Customers              : "
    f"{total_customers:,}"
)


print(
    f"Products               : "
    f"{total_products:,}"
)


print(
    f"Units Sold             : "
    f"{total_units:,.0f}"
)


print(
    f"Average Order Value     : "
    f"{average_order_value:,.2f}"
)


print(
    f"Revenue / Customer      : "
    f"{revenue_per_customer:,.2f}"
)


print(
    f"Invoice Cancellation % : "
    f"{invoice_cancellation_rate:.2f}%"
)


print(
    f"Top 20% Customer Share : "
    f"{top_20_customer_revenue_pct:.2f}%"
)


print(
    f"Top 20% Product Share  : "
    f"{top_20_product_revenue_pct:.2f}%"
)


print(
    "=" * 60
)


print(
    "\nNotebook 06 completed successfully."
)


```
