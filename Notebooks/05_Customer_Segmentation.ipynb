```python
# ============================================================
# 05_Customer_Segmentation.ipynb
# Online Retail Customer Analytics
#
# Customer Segmentation Using Behavioural Features,
# K-Means Clustering and PCA
# ============================================================


# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

# Data manipulation
import pandas as pd
import numpy as np

# Visualisation
import matplotlib.pyplot as plt
import seaborn as sns

# Preprocessing
from sklearn.preprocessing import StandardScaler

# Clustering
from sklearn.cluster import KMeans

# Dimensionality reduction
from sklearn.decomposition import PCA

# Cluster evaluation
from sklearn.metrics import (
    silhouette_score,
    davies_bouldin_score,
    calinski_harabasz_score
)

# Model persistence
import joblib

# File / directory management
from pathlib import Path


# Display settings
pd.set_option(
    "display.max_columns",
    None
)

pd.set_option(
    "display.float_format",
    lambda x: f"{x:,.2f}"
)


print("Libraries imported successfully.")



# ============================================================
# 2. LOAD CUSTOMER FEATURE DATASET
# ============================================================

customer_df = pd.read_csv(
    "../Data/customer_features.csv"
)


print(
    "Customer dataset loaded successfully."
)

print(
    "Dataset Shape:",
    customer_df.shape
)

display(
    customer_df.head()
)



# ============================================================
# 3. SEGMENTATION DATASET VALIDATION
# ============================================================


# ------------------------------------------------------------
# 3.1 Dataset Structure
# ------------------------------------------------------------

print(
    "Dataset Shape:",
    customer_df.shape
)

print(
    "\nDataset Information:"
)

customer_df.info()



# ------------------------------------------------------------
# 3.2 Available Features
# ------------------------------------------------------------

print(
    "\nCustomer Dataset Columns:"
)

for number, column in enumerate(
    customer_df.columns,
    start=1
):

    print(
        f"{number}. {column}"
    )



# ------------------------------------------------------------
# 3.3 Missing Values
# ------------------------------------------------------------

missing_summary = pd.DataFrame({

    "Missing_Count":
        customer_df.isna().sum(),

    "Missing_Percentage":
        (
            customer_df.isna().mean()
            * 100
        ).round(2)
})


display(
    missing_summary[
        missing_summary[
            "Missing_Count"
        ] > 0
    ]
)


print(
    "Total Missing Values:",
    customer_df.isna()
    .sum()
    .sum()
)



# ------------------------------------------------------------
# 3.4 Duplicate Validation
# ------------------------------------------------------------

print(
    "Exact Duplicate Rows:",
    customer_df
    .duplicated()
    .sum()
)


if "CustomerID" in customer_df.columns:

    print(
        "Duplicate Customer IDs:",
        customer_df[
            "CustomerID"
        ]
        .duplicated()
        .sum()
    )



# ------------------------------------------------------------
# 3.5 Numerical Feature Identification
# ------------------------------------------------------------

numeric_columns = (
    customer_df
    .select_dtypes(
        include=np.number
    )
    .columns
    .tolist()
)


print(
    "\nNumerical Customer Features:"
)

print(
    numeric_columns
)



# ------------------------------------------------------------
# 3.6 Infinite Value Validation
# ------------------------------------------------------------

numeric_df = customer_df[
    numeric_columns
]


infinity_summary = pd.DataFrame({

    "Positive_Infinity":
        np.isposinf(
            numeric_df
        ).sum(),

    "Negative_Infinity":
        np.isneginf(
            numeric_df
        ).sum(),

    "Total_Infinite":
        np.isinf(
            numeric_df
        ).sum()
})


display(
    infinity_summary
)



# ------------------------------------------------------------
# 3.7 Descriptive Statistics
# ------------------------------------------------------------

display(
    customer_df[
        numeric_columns
    ]
    .describe()
    .T
)



# ============================================================
# 4. SEGMENTATION FEATURE SELECTION
# ============================================================


# ------------------------------------------------------------
# 4.1 Separate Identifier Features
# ------------------------------------------------------------

identifier_features = [
    column
    for column in [
        "CustomerID"
    ]
    if column in customer_df.columns
]


print(
    "Identifier Features:"
)

print(
    identifier_features
)



# ------------------------------------------------------------
# 4.2 Candidate Behavioural Features
# ------------------------------------------------------------

# Candidate variables representing customer behaviour.
#
# Only features that actually exist in customer_features.csv
# will be retained.

candidate_feature_names = [

    # RFM-style features
    "Recency",
    "Frequency",
    "Monetary",

    # Customer value
    "TotalRevenue",
    "AverageOrderValue",

    # Purchasing behaviour
    "OrderCount",
    "TotalQuantity",
    "UniqueProducts",

    # Possible additional engineered features
    "CustomerTenure",
    "TransactionCount"
]


candidate_features = [

    feature

    for feature
    in candidate_feature_names

    if feature
    in customer_df.columns
]


print(
    "\nAvailable Candidate Segmentation Features:"
)

for feature in candidate_features:

    print(
        feature
    )



# ------------------------------------------------------------
# 4.3 Candidate Feature Descriptive Statistics
# ------------------------------------------------------------

if len(
    candidate_features
) > 0:

    display(

        customer_df[
            candidate_features
        ]
        .describe()
        .T
    )



# ------------------------------------------------------------
# 4.4 Candidate Feature Correlation
# ------------------------------------------------------------

candidate_correlation = (

    customer_df[
        candidate_features
    ]
    .corr(
        method="spearman"
    )
)


display(
    candidate_correlation.round(2)
)



# ------------------------------------------------------------
# 4.5 Candidate Feature Correlation Heatmap
# ------------------------------------------------------------

if len(
    candidate_features
) >= 2:

    fig, ax = plt.subplots(
        figsize=(10, 7)
    )

    sns.heatmap(

        candidate_correlation,

        annot=True,

        fmt=".2f",

        center=0,

        ax=ax
    )

    ax.set_title(
        "Spearman Correlation Between Candidate Segmentation Features"
    )

    ax.set_xlabel(
        "Customer Features"
    )

    ax.set_ylabel(
        "Customer Features"
    )

    plt.tight_layout()

    plt.show()



# ------------------------------------------------------------
# 4.6 Define Final Clustering Features
# ------------------------------------------------------------

# IMPORTANT:
#
# RFM variables should be preferred where they exist because
# they provide clear and interpretable customer dimensions.
#
# Avoid including exact conceptual duplicates:
#
# Frequency + OrderCount
# Monetary + TotalRevenue
#
# The following logic selects one representation from each
# major behavioural dimension.


final_clustering_features = []


# RECENCY
if "Recency" in customer_df.columns:

    final_clustering_features.append(
        "Recency"
    )


# FREQUENCY
if "Frequency" in customer_df.columns:

    final_clustering_features.append(
        "Frequency"
    )

elif "OrderCount" in customer_df.columns:

    final_clustering_features.append(
        "OrderCount"
    )


# MONETARY VALUE
if "Monetary" in customer_df.columns:

    final_clustering_features.append(
        "Monetary"
    )

elif "TotalRevenue" in customer_df.columns:

    final_clustering_features.append(
        "TotalRevenue"
    )


# PRODUCT DIVERSITY
if "UniqueProducts" in customer_df.columns:

    final_clustering_features.append(
        "UniqueProducts"
    )


# If fewer than 3 dimensions are available, include
# additional behavioural variables.

additional_candidates = [

    "AverageOrderValue",
    "TotalQuantity",
    "CustomerTenure"
]


for feature in additional_candidates:

    if (
        len(
            final_clustering_features
        ) < 4
        and
        feature
        in customer_df.columns
        and
        feature
        not in final_clustering_features
    ):

        final_clustering_features.append(
            feature
        )


print(
    "\nFinal Clustering Features:"
)

for feature in final_clustering_features:

    print(
        feature
    )


if len(
    final_clustering_features
) < 2:

    raise ValueError(
        "Insufficient customer behavioural features "
        "were found for clustering."
    )



# ------------------------------------------------------------
# 4.7 Final Feature Correlation Check
# ------------------------------------------------------------

final_feature_correlation = (

    customer_df[
        final_clustering_features
    ]
    .corr(
        method="spearman"
    )
)


display(
    final_feature_correlation.round(2)
)



# ============================================================
# 5. DISTRIBUTION AND SKEWNESS ANALYSIS
# ============================================================


# ------------------------------------------------------------
# 5.1 Create Segmentation Working Dataset
# ------------------------------------------------------------

segmentation_df = (

    customer_df[
        final_clustering_features
    ]
    .copy()
)


print(
    "Segmentation Dataset Shape:",
    segmentation_df.shape
)


display(
    segmentation_df.head()
)



# ------------------------------------------------------------
# 5.2 Final Missing / Infinity Check
# ------------------------------------------------------------

print(
    "Missing Values:",
    segmentation_df
    .isna()
    .sum()
    .sum()
)


print(
    "Infinite Values:",
    np.isinf(
        segmentation_df
    )
    .sum()
    .sum()
)



# ------------------------------------------------------------
# 5.3 Skewness Analysis
# ------------------------------------------------------------

skewness_summary = pd.DataFrame({

    "Feature":
        final_clustering_features,

    "Skewness": [

        segmentation_df[
            feature
        ].skew()

        for feature
        in final_clustering_features
    ]
})


skewness_summary[
    "Absolute_Skewness"
] = (

    skewness_summary[
        "Skewness"
    ].abs()
)


skewness_summary = (

    skewness_summary
    .sort_values(

        "Absolute_Skewness",

        ascending=False
    )
    .reset_index(
        drop=True
    )
)


display(
    skewness_summary.round(3)
)



# ------------------------------------------------------------
# 5.4 Distribution Analysis
# ------------------------------------------------------------

for feature in final_clustering_features:

    fig, ax = plt.subplots(
        figsize=(9, 5)
    )

    sns.histplot(

        data=segmentation_df,

        x=feature,

        bins=40,

        kde=True,

        ax=ax
    )

    ax.set_title(
        f"Distribution of {feature}"
    )

    ax.set_xlabel(
        feature
    )

    ax.set_ylabel(
        "Frequency"
    )

    plt.tight_layout()

    plt.show()



# ------------------------------------------------------------
# 5.5 Boxplot Analysis
# ------------------------------------------------------------

for feature in final_clustering_features:

    fig, ax = plt.subplots(
        figsize=(9, 4)
    )

    sns.boxplot(

        data=segmentation_df,

        x=feature,

        ax=ax
    )

    ax.set_title(
        f"Boxplot of {feature}"
    )

    ax.set_xlabel(
        feature
    )

    plt.tight_layout()

    plt.show()



# ------------------------------------------------------------
# 5.6 IQR Extreme-Value Summary
# ------------------------------------------------------------

outlier_results = []


for feature in final_clustering_features:

    series = (

        segmentation_df[
            feature
        ]
        .replace(
            [np.inf, -np.inf],
            np.nan
        )
        .dropna()
    )


    q1 = series.quantile(
        0.25
    )

    q3 = series.quantile(
        0.75
    )

    iqr = q3 - q1


    lower_bound = (
        q1
        -
        1.5 * iqr
    )

    upper_bound = (
        q3
        +
        1.5 * iqr
    )


    extreme_mask = (

        (series < lower_bound)

        |

        (series > upper_bound)
    )


    extreme_count = (
        extreme_mask.sum()
    )


    extreme_percentage = (

        extreme_count

        /

        len(series)

        * 100
    )


    outlier_results.append({

        "Feature":
            feature,

        "Q1":
            q1,

        "Q3":
            q3,

        "IQR":
            iqr,

        "Lower_Bound":
            lower_bound,

        "Upper_Bound":
            upper_bound,

        "Extreme_Count":
            extreme_count,

        "Extreme_Percentage":
            extreme_percentage
    })


outlier_summary = pd.DataFrame(
    outlier_results
)


display(
    outlier_summary.round(2)
)



# ============================================================
# 6. FEATURE TRANSFORMATION
# ============================================================


# ------------------------------------------------------------
# 6.1 Create Transformed Dataset
# ------------------------------------------------------------

segmentation_transformed = (
    segmentation_df.copy()
)



# ------------------------------------------------------------
# 6.2 Identify Strongly Right-Skewed Non-Negative Features
# ------------------------------------------------------------

log_transform_features = []


for feature in final_clustering_features:

    feature_skewness = (

        segmentation_df[
            feature
        ].skew()
    )


    minimum_value = (

        segmentation_df[
            feature
        ].min()
    )


    if (
        feature_skewness > 1
        and
        minimum_value >= 0
    ):

        log_transform_features.append(
            feature
        )


print(
    "Features Selected for log1p Transformation:"
)

print(
    log_transform_features
)



# ------------------------------------------------------------
# 6.3 Apply log1p Transformation
# ------------------------------------------------------------

for feature in log_transform_features:

    segmentation_transformed[
        feature
    ] = np.log1p(

        segmentation_transformed[
            feature
        ]
    )



# ------------------------------------------------------------
# 6.4 Compare Skewness Before and After
# ------------------------------------------------------------

transformation_comparison = pd.DataFrame({

    "Original_Skewness":

        segmentation_df[
            final_clustering_features
        ].skew(),

    "Transformed_Skewness":

        segmentation_transformed[
            final_clustering_features
        ].skew()
})


transformation_comparison[
    "Absolute_Improvement"
] = (

    transformation_comparison[
        "Original_Skewness"
    ].abs()

    -

    transformation_comparison[
        "Transformed_Skewness"
    ].abs()
)


display(
    transformation_comparison.round(3)
)



# ------------------------------------------------------------
# 6.5 Visualise Transformed Features
# ------------------------------------------------------------

for feature in log_transform_features:

    fig, ax = plt.subplots(
        figsize=(9, 5)
    )

    sns.histplot(

        data=segmentation_transformed,

        x=feature,

        bins=40,

        kde=True,

        ax=ax
    )

    ax.set_title(
        f"Transformed Distribution of {feature}"
    )

    ax.set_xlabel(
        f"log1p({feature})"
    )

    ax.set_ylabel(
        "Frequency"
    )

    plt.tight_layout()

    plt.show()



# ============================================================
# 7. FEATURE SCALING
# ============================================================


# ------------------------------------------------------------
# 7.1 Initialise StandardScaler
# ------------------------------------------------------------

scaler = StandardScaler()



# ------------------------------------------------------------
# 7.2 Fit and Transform Features
# ------------------------------------------------------------

scaled_features = (

    scaler.fit_transform(

        segmentation_transformed[
            final_clustering_features
        ]
    )
)



# ------------------------------------------------------------
# 7.3 Convert Scaled Data to DataFrame
# ------------------------------------------------------------

segmentation_scaled = pd.DataFrame(

    scaled_features,

    columns=final_clustering_features,

    index=segmentation_transformed.index
)


print(
    "Scaled Dataset Shape:",
    segmentation_scaled.shape
)


display(
    segmentation_scaled.head()
)



# ------------------------------------------------------------
# 7.4 Validate Scaling
# ------------------------------------------------------------

scaling_validation = pd.DataFrame({

    "Mean":
        segmentation_scaled.mean(),

    "Standard_Deviation":
        segmentation_scaled.std(
            ddof=0
        ),

    "Minimum":
        segmentation_scaled.min(),

    "Maximum":
        segmentation_scaled.max()
})


display(
    scaling_validation.round(3)
)



# ============================================================
# 8. DETERMINE OPTIMAL NUMBER OF CLUSTERS
# ============================================================


# ------------------------------------------------------------
# 8.1 Test Candidate K Values
# ------------------------------------------------------------

k_values = range(
    2,
    11
)


inertia_scores = []

silhouette_scores = []

davies_bouldin_scores = []

calinski_harabasz_scores = []



for k in k_values:

    kmeans = KMeans(

        n_clusters=k,

        random_state=42,

        n_init=10
    )


    labels = kmeans.fit_predict(
        segmentation_scaled
    )


    inertia_scores.append(
        kmeans.inertia_
    )


    silhouette_scores.append(

        silhouette_score(

            segmentation_scaled,

            labels
        )
    )


    davies_bouldin_scores.append(

        davies_bouldin_score(

            segmentation_scaled,

            labels
        )
    )


    calinski_harabasz_scores.append(

        calinski_harabasz_score(

            segmentation_scaled,

            labels
        )
    )



# ------------------------------------------------------------
# 8.2 Cluster Evaluation Table
# ------------------------------------------------------------

cluster_evaluation = pd.DataFrame({

    "K":
        list(k_values),

    "Inertia":
        inertia_scores,

    "Silhouette_Score":
        silhouette_scores,

    "Davies_Bouldin_Score":
        davies_bouldin_scores,

    "Calinski_Harabasz_Score":
        calinski_harabasz_scores
})


display(
    cluster_evaluation.round(3)
)



# ------------------------------------------------------------
# 8.3 Elbow Method
# ------------------------------------------------------------

fig, ax = plt.subplots(
    figsize=(9, 5)
)


ax.plot(

    list(k_values),

    inertia_scores,

    marker="o"
)


ax.set_title(
    "Elbow Method for K-Means Clustering"
)

ax.set_xlabel(
    "Number of Clusters (K)"
)

ax.set_ylabel(
    "Inertia"
)

ax.set_xticks(
    list(k_values)
)


plt.tight_layout()

plt.show()



# ------------------------------------------------------------
# 8.4 Silhouette Score
# ------------------------------------------------------------

fig, ax = plt.subplots(
    figsize=(9, 5)
)


ax.plot(

    list(k_values),

    silhouette_scores,

    marker="o"
)


ax.set_title(
    "Silhouette Score by Number of Clusters"
)

ax.set_xlabel(
    "Number of Clusters (K)"
)

ax.set_ylabel(
    "Silhouette Score"
)

ax.set_xticks(
    list(k_values)
)


plt.tight_layout()

plt.show()



# ------------------------------------------------------------
# 8.5 Davies-Bouldin Index
# ------------------------------------------------------------

fig, ax = plt.subplots(
    figsize=(9, 5)
)


ax.plot(

    list(k_values),

    davies_bouldin_scores,

    marker="o"
)


ax.set_title(
    "Davies-Bouldin Index by Number of Clusters"
)

ax.set_xlabel(
    "Number of Clusters (K)"
)

ax.set_ylabel(
    "Davies-Bouldin Index"
)

ax.set_xticks(
    list(k_values)
)


plt.tight_layout()

plt.show()



# ------------------------------------------------------------
# 8.6 Calinski-Harabasz Score
# ------------------------------------------------------------

fig, ax = plt.subplots(
    figsize=(9, 5)
)


ax.plot(

    list(k_values),

    calinski_harabasz_scores,

    marker="o"
)


ax.set_title(
    "Calinski-Harabasz Score by Number of Clusters"
)

ax.set_xlabel(
    "Number of Clusters (K)"
)

ax.set_ylabel(
    "Calinski-Harabasz Score"
)

ax.set_xticks(
    list(k_values)
)


plt.tight_layout()

plt.show()



# ============================================================
# 9. SELECT FINAL K
# ============================================================

# IMPORTANT:
#
# Review:
#
# 1. Elbow Method
# 2. Silhouette Score
# 3. Davies-Bouldin Index
# 4. Calinski-Harabasz Score
# 5. Business interpretability
# 6. Cluster size balance
#
# Then manually set the selected K below.


optimal_k = 4


print(
    "Selected Number of Clusters:",
    optimal_k
)



# ============================================================
# 10. TRAIN FINAL K-MEANS MODEL
# ============================================================


final_kmeans = KMeans(

    n_clusters=optimal_k,

    random_state=42,

    n_init=10
)


cluster_labels = (

    final_kmeans.fit_predict(
        segmentation_scaled
    )
)


print(
    "Final K-Means model trained successfully."
)



# ============================================================
# 11. ASSIGN CLUSTER LABELS
# ============================================================


customer_segmented_df = (
    customer_df.copy()
)


customer_segmented_df[
    "Cluster"
] = cluster_labels


display(
    customer_segmented_df.head()
)



# ------------------------------------------------------------
# 11.1 Cluster Distribution
# ------------------------------------------------------------

cluster_sizes = (

    customer_segmented_df[
        "Cluster"
    ]
    .value_counts()
    .sort_index()
)


cluster_percentages = (

    customer_segmented_df[
        "Cluster"
    ]
    .value_counts(
        normalize=True
    )
    .sort_index()

    * 100
)


cluster_distribution = pd.DataFrame({

    "Customer_Count":
        cluster_sizes,

    "Percentage":
        cluster_percentages
})


display(
    cluster_distribution.round(2)
)



# ------------------------------------------------------------
# 11.2 Cluster Distribution Visualisation
# ------------------------------------------------------------

fig, ax = plt.subplots(
    figsize=(8, 5)
)


sns.countplot(

    data=customer_segmented_df,

    x="Cluster",

    ax=ax
)


ax.set_title(
    "Customer Distribution Across K-Means Clusters"
)

ax.set_xlabel(
    "Cluster"
)

ax.set_ylabel(
    "Number of Customers"
)


plt.tight_layout()

plt.show()



# ============================================================
# 12. FINAL CLUSTER EVALUATION
# ============================================================


final_inertia = (
    final_kmeans.inertia_
)


final_silhouette = (

    silhouette_score(

        segmentation_scaled,

        cluster_labels
    )
)


final_davies_bouldin = (

    davies_bouldin_score(

        segmentation_scaled,

        cluster_labels
    )
)


final_calinski_harabasz = (

    calinski_harabasz_score(

        segmentation_scaled,

        cluster_labels
    )
)



final_evaluation = pd.DataFrame({

    "Metric": [

        "Inertia",

        "Silhouette Score",

        "Davies-Bouldin Index",

        "Calinski-Harabasz Score"
    ],

    "Value": [

        final_inertia,

        final_silhouette,

        final_davies_bouldin,

        final_calinski_harabasz
    ]
})


display(
    final_evaluation.round(3)
)



# ============================================================
# 13. CLUSTER PROFILING
# ============================================================


# ------------------------------------------------------------
# 13.1 Mean Profile
# ------------------------------------------------------------

cluster_means = (

    customer_segmented_df

    .groupby(
        "Cluster"
    )[
        final_clustering_features
    ]

    .mean()
)


print(
    "Cluster Mean Profiles:"
)


display(
    cluster_means.round(2)
)



# ------------------------------------------------------------
# 13.2 Median Profile
# ------------------------------------------------------------

cluster_medians = (

    customer_segmented_df

    .groupby(
        "Cluster"
    )[
        final_clustering_features
    ]

    .median()
)


print(
    "Cluster Median Profiles:"
)


display(
    cluster_medians.round(2)
)



# ------------------------------------------------------------
# 13.3 Mean and Median Combined Profile
# ------------------------------------------------------------

cluster_profile = (

    customer_segmented_df

    .groupby(
        "Cluster"
    )[
        final_clustering_features
    ]

    .agg(
        [
            "mean",
            "median"
        ]
    )
)


display(
    cluster_profile.round(2)
)



# ------------------------------------------------------------
# 13.4 Relative Cluster Profile
# ------------------------------------------------------------

overall_means = (

    customer_segmented_df[
        final_clustering_features
    ]
    .mean()
)


cluster_relative_profile = (

    cluster_means
    .divide(
        overall_means
    )
)


display(
    cluster_relative_profile.round(2)
)



# ------------------------------------------------------------
# 13.5 Relative Profile Heatmap
# ------------------------------------------------------------

fig, ax = plt.subplots(
    figsize=(10, 6)
)


sns.heatmap(

    cluster_relative_profile,

    annot=True,

    fmt=".2f",

    center=1,

    ax=ax
)


ax.set_title(
    "Relative Customer Behaviour by Cluster"
)

ax.set_xlabel(
    "Customer Behaviour Features"
)

ax.set_ylabel(
    "Cluster"
)


plt.tight_layout()

plt.show()



# ------------------------------------------------------------
# 13.6 Extended Business Profile
# ------------------------------------------------------------

profiling_candidates = [

    "Recency",

    "Frequency",

    "Monetary",

    "TotalRevenue",

    "OrderCount",

    "AverageOrderValue",

    "TotalQuantity",

    "UniqueProducts",

    "CustomerTenure",

    "TransactionCount"
]


profiling_features = [

    feature

    for feature
    in profiling_candidates

    if feature
    in customer_segmented_df.columns
]


cluster_business_profile = (

    customer_segmented_df

    .groupby(
        "Cluster"
    )[
        profiling_features
    ]

    .mean()
)


print(
    "Extended Cluster Business Profile:"
)


display(
    cluster_business_profile.round(2)
)



# ============================================================
# 14. PCA VISUALISATION
# ============================================================


# ------------------------------------------------------------
# 14.1 Apply PCA
# ------------------------------------------------------------

pca = PCA(
    n_components=2
)


pca_components = (

    pca.fit_transform(
        segmentation_scaled
    )
)


print(
    "PCA Explained Variance Ratio:"
)

print(
    pca.explained_variance_ratio_
)


print(
    "Total Explained Variance:"
)

print(
    pca.explained_variance_ratio_
    .sum()
)



# ------------------------------------------------------------
# 14.2 Create PCA DataFrame
# ------------------------------------------------------------

pca_df = pd.DataFrame(

    pca_components,

    columns=[
        "PC1",
        "PC2"
    ],

    index=segmentation_scaled.index
)


pca_df[
    "Cluster"
] = cluster_labels


display(
    pca_df.head()
)



# ------------------------------------------------------------
# 14.3 PCA Cluster Scatterplot
# ------------------------------------------------------------

fig, ax = plt.subplots(
    figsize=(10, 7)
)


sns.scatterplot(

    data=pca_df,

    x="PC1",

    y="PC2",

    hue="Cluster",

    palette="tab10",

    alpha=0.65,

    s=60,

    ax=ax
)


ax.set_title(
    "Customer Segments Visualised Using PCA"
)


ax.set_xlabel(

    f"Principal Component 1 "
    f"({pca.explained_variance_ratio_[0] * 100:.1f}% variance)"
)


ax.set_ylabel(

    f"Principal Component 2 "
    f"({pca.explained_variance_ratio_[1] * 100:.1f}% variance)"
)


ax.legend(
    title="Cluster"
)


plt.tight_layout()

plt.show()



# ------------------------------------------------------------
# 14.4 PCA Feature Loadings
# ------------------------------------------------------------

pca_loadings = pd.DataFrame(

    pca.components_.T,

    columns=[
        "PC1",
        "PC2"
    ],

    index=final_clustering_features
)


display(
    pca_loadings.round(3)
)



# ------------------------------------------------------------
# 14.5 PCA Loading Heatmap
# ------------------------------------------------------------

fig, ax = plt.subplots(
    figsize=(8, 6)
)


sns.heatmap(

    pca_loadings,

    annot=True,

    fmt=".2f",

    center=0,

    ax=ax
)


ax.set_title(
    "PCA Feature Loadings"
)

ax.set_xlabel(
    "Principal Components"
)

ax.set_ylabel(
    "Customer Features"
)


plt.tight_layout()

plt.show()



# ============================================================
# 15. BUSINESS SEGMENT INTERPRETATION
# ============================================================


# ------------------------------------------------------------
# 15.1 Create Interpretation Table
# ------------------------------------------------------------

cluster_interpretation = (
    cluster_business_profile.copy()
)


customer_count = (

    customer_segmented_df

    .groupby(
        "Cluster"
    )

    .size()
)


customer_percentage = (

    customer_count

    /

    len(
        customer_segmented_df
    )

    * 100
)


cluster_interpretation.insert(

    0,

    "CustomerCount",

    customer_count
)


cluster_interpretation.insert(

    1,

    "CustomerPercentage",

    customer_percentage
)


display(
    cluster_interpretation.round(2)
)



# ============================================================
# 16. ASSIGN BUSINESS-FRIENDLY SEGMENT NAMES
# ============================================================

# IMPORTANT:
#
# Do NOT assign names until the actual cluster profiles above
# have been interpreted.
#
# Replace the placeholder labels below with names supported by
# your cluster results.
#
# Possible examples ONLY:
#
# High-Value Loyal Customers
# Regular Customers
# New / Developing Customers
# Inactive / At-Risk Customers
#
# Cluster numbers themselves have no inherent meaning.


segment_names = {

    cluster:
        f"Segment {cluster}"

    for cluster
    in sorted(
        customer_segmented_df[
            "Cluster"
        ].unique()
    )
}


# ------------------------------------------------------------
# AFTER interpreting your results, replace the dictionary above
# with something similar to:
#
# segment_names = {
#     0: "High-Value Loyal",
#     1: "At-Risk",
#     2: "Developing",
#     3: "Regular"
# }
#
# ONLY if your actual cluster profiles support those labels.
# ------------------------------------------------------------


customer_segmented_df[
    "Segment"
] = (

    customer_segmented_df[
        "Cluster"
    ]
    .map(
        segment_names
    )
)


display(
    customer_segmented_df.head()
)



# ============================================================
# 17. SEGMENT-LEVEL BUSINESS ANALYSIS
# ============================================================


# ------------------------------------------------------------
# 17.1 Segment Summary
# ------------------------------------------------------------

segment_summary = (

    customer_segmented_df

    .groupby(
        "Segment"
    )

    .agg(

        CustomerCount=(
            "Cluster",
            "size"
        ),

        **{

            f"Mean_{feature}":
                (
                    feature,
                    "mean"
                )

            for feature
            in profiling_features
        }
    )
)


segment_summary[
    "CustomerPercentage"
] = (

    segment_summary[
        "CustomerCount"
    ]

    /

    len(
        customer_segmented_df
    )

    * 100
)


display(
    segment_summary.round(2)
)



# ------------------------------------------------------------
# 17.2 Segment Size Visualisation
# ------------------------------------------------------------

segment_counts = (

    customer_segmented_df[
        "Segment"
    ]
    .value_counts()
)


fig, ax = plt.subplots(
    figsize=(9, 5)
)


ax.bar(

    segment_counts.index,

    segment_counts.values
)


ax.set_title(
    "Customer Segment Distribution"
)

ax.set_xlabel(
    "Customer Segment"
)

ax.set_ylabel(
    "Number of Customers"
)


plt.xticks(
    rotation=30,
    ha="right"
)


plt.tight_layout()

plt.show()



# ------------------------------------------------------------
# 17.3 Revenue Contribution by Segment
# ------------------------------------------------------------

revenue_column = None


if "Monetary" in customer_segmented_df.columns:

    revenue_column = "Monetary"

elif "TotalRevenue" in customer_segmented_df.columns:

    revenue_column = "TotalRevenue"


if revenue_column is not None:

    segment_revenue = (

        customer_segmented_df

        .groupby(
            "Segment"
        )[
            revenue_column
        ]

        .sum()

        .sort_values(
            ascending=False
        )
    )


    segment_revenue_percentage = (

        segment_revenue

        /

        segment_revenue.sum()

        * 100
    )


    revenue_contribution = pd.DataFrame({

        "Revenue":
            segment_revenue,

        "Revenue_Percentage":
            segment_revenue_percentage
    })


    display(
        revenue_contribution.round(2)
    )



# ============================================================
# 18. EXPORT SEGMENTATION RESULTS
# ============================================================


# ------------------------------------------------------------
# 18.1 Create Output Directories
# ------------------------------------------------------------

Path(
    "../Data"
).mkdir(
    parents=True,
    exist_ok=True
)


Path(
    "../models"
).mkdir(
    parents=True,
    exist_ok=True
)



# ------------------------------------------------------------
# 18.2 Export Customer Segments
# ------------------------------------------------------------

customer_segmented_df.to_csv(

    "../Data/customer_segments.csv",

    index=False
)


print(
    "customer_segments.csv saved successfully."
)



# ------------------------------------------------------------
# 18.3 Export Cluster Evaluation
# ------------------------------------------------------------

cluster_evaluation.to_csv(

    "../Data/cluster_evaluation.csv",

    index=False
)



# ------------------------------------------------------------
# 18.4 Export Cluster Business Profile
# ------------------------------------------------------------

cluster_business_profile.to_csv(

    "../Data/cluster_business_profile.csv"
)



# ------------------------------------------------------------
# 18.5 Export PCA Coordinates
# ------------------------------------------------------------

pca_export = pca_df.copy()


if "CustomerID" in customer_segmented_df.columns:

    pca_export.insert(

        0,

        "CustomerID",

        customer_segmented_df[
            "CustomerID"
        ].values
    )


pca_export.to_csv(

    "../Data/customer_pca.csv",

    index=False
)



# ============================================================
# 19. SAVE MODEL AND PREPROCESSING OBJECTS
# ============================================================


# ------------------------------------------------------------
# 19.1 Save StandardScaler
# ------------------------------------------------------------

joblib.dump(

    scaler,

    "../models/customer_scaler.pkl"
)



# ------------------------------------------------------------
# 19.2 Save K-Means Model
# ------------------------------------------------------------

joblib.dump(

    final_kmeans,

    "../models/kmeans_model.pkl"
)



# ------------------------------------------------------------
# 19.3 Save PCA Model
# ------------------------------------------------------------

joblib.dump(

    pca,

    "../models/customer_pca.pkl"
)



# ------------------------------------------------------------
# 19.4 Save Preprocessing Metadata
# ------------------------------------------------------------

preprocessing_metadata = {

    "clustering_features":
        final_clustering_features,

    "log_transformed_features":
        log_transform_features,

    "optimal_k":
        optimal_k
}


joblib.dump(

    preprocessing_metadata,

    "../models/segmentation_metadata.pkl"
)


print(
    "Clustering models and preprocessing metadata saved successfully."
)



# ============================================================
# 20. FINAL EXPORT VALIDATION
# ============================================================


segment_check = pd.read_csv(
    "../Data/customer_segments.csv"
)


print(
    "Exported Customer Segment Dataset Shape:",
    segment_check.shape
)


display(
    segment_check.head()
)


print(
    "\nExported Segment Distribution:"
)


display(

    segment_check[
        "Segment"
    ]
    .value_counts()
    .to_frame(
        "Customer_Count"
    )
)



# ============================================================
# 21. FINAL NOTEBOOK SUMMARY
# ============================================================


final_summary = pd.DataFrame({

    "Item": [

        "Customers Analysed",

        "Clustering Features",

        "Log-Transformed Features",

        "Selected K",

        "Silhouette Score",

        "Davies-Bouldin Index",

        "Calinski-Harabasz Score",

        "PCA Explained Variance"
    ],

    "Value": [

        len(
            customer_segmented_df
        ),

        ", ".join(
            final_clustering_features
        ),

        (
            ", ".join(
                log_transform_features
            )

            if len(
                log_transform_features
            ) > 0

            else "None"
        ),

        optimal_k,

        round(
            final_silhouette,
            3
        ),

        round(
            final_davies_bouldin,
            3
        ),

        round(
            final_calinski_harabasz,
            3
        ),

        round(
            pca.explained_variance_ratio_
            .sum(),
            3
        )
    ]
})


display(
    final_summary
)


print(
    "\nCustomer segmentation notebook completed successfully."
)
```
